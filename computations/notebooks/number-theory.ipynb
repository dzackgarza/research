## Point counts over finite fields and the Weil conjectures

```python
%display latex
# https://www.projecteuclid.org/download/pdf_1/euclid.bams/1183513798
```

```python
K = GF(5)
X = ProjectiveSpaces(K)(2, names=("x0", "x1", "x2"))
n = [3, 3, 3]
a = [1, 1, 1]
q = 3
d = [ZZ(ni).gcd(q - 1) for ni in n]
print(d)
```

```python
pts = list(X.rational_points())
print(len(pts))
pts[0:10]
```

```python
# Defines the variety we want to count points on
def f(x):
    return (a[0] * x[0]^n[0]) + (a[1] * x[1]^n[1]) + (a[2] * x[2]^n[2])

# "Reduced" version of f
def L(u):
    return (a[0] * u[0]) + (a[1] * u[1]) + (a[2] * u[2])

def nth_roots(u):
    return(u.nth_root(n[0], all=True))

# Number of solutions to x^{n_i} in K
def Ni(ui): 
    return(1 if ui == 0 else len(nth_roots(ui)))
```

```python
X_solns = [x for x in pts if f(x) == 0]
L_solns = [u for u in pts if L(u) == 0]
show(X_solns)
show(L_solns)
```

```python
[f(x) for x in pts]
```

```python
show(len(X_solns))
show(len(L_solns))
```

```python
Nis = [(Ni(u[0]), Ni(u[1]), Ni(u[2])) for u in L_solns]
show(Nis)
sum([ni[0] * ni[1] * ni[2] for ni in Nis])
```

## Weil conjectures: counting points on affine varieties

[Link to Weil Paper](https://www.projecteuclid.org/download/pdf_1/euclid.bams/1183513798)

```python
%display latex
```

```python
q = 5
K = GF(q)
A = AffineSpaces(K)(3)

def f(x):
    return(x[0]^2 + x[1]^2 + x[2]^2)
```

```python
A.origin()
```

Generally, we're fixing a field $k$ of positive characteristic, considering equations of the form
$$
\sum_{i=1}^n a_i x_i^{n_i} = 0
$$
in the polynomial ring $A = k[x_1, \cdots, x_n]$.



For an example, take $K = \mathbb{F}_5$ and $A = k[x_1, x_2]$, and consider the variety
$$
f(x_1, x_2) = x_1^2 + x_2^2 = 0
$$

where 

- $q = 2$
- $n_i = 2$ for all $i$
- $a_i = 1$ for all $i$
- $N = 8$ (as we will find out)

```python
A_pts = [p for p in A.rational_points() if p != A.origin()]
A_pts
```

```python
#for p in A_pts:
#    show("f", p, "=", f(p))
```

```python
# Brute force all of the solutions
solutions = [(p, f(p)) for p in A_pts if f(p) == 0]

# List them out
for s in solutions: 
    show(s[0][0], "^2 +", s[0][1], "^2 = ", s[1])
```

```python
# Count them
N = len(solutions)
show(N)
```

```python
di = ZZ(2).gcd(q - 1)
show("q=",q)
show("d_i=",di)
```

```python
def Ni(u):
    ni_th_roots = u.nth_root(2, all=True)
    return(len(ni_th_roots))
```

```python
[(u, Ni(u)) for u in K]
```

```python
# Compute all of the squares in $K$.
[(x, x^2) for x in K]
```

We can also check manually:

- 0 has one square root: 0
- 1 has one square root: 1
- 2 has zero square roots
- 3 has zero square roots
- 4 has one square root: 3

Weyl now says that we should consider the auxiliary equation
$$
L(u) = \sum_{i=1}^n a_i x_i
$$
obtained by dropping all exponents.

```python
def L(x):
    return(x[0] + x[1] + x[2])

L_solns = [p for p in A_pts if L(p) == 0]
show(L_solns)
show(len(L_solns))
```

```python
sum([(Ni(p[0]) * Ni(p[1])) for p in L_solns])
```

## Distribution of sums of two squares

```python
from collections import Counter
%matplotlib notebook

import matplotlib
import numpy as np
import matplotlib.pyplot as plt
n = 20
```

```python
s = Set()
Qring = ZZ.polynomial_ring(("x", "y"))
x, y = Qring.algebra_generator("x"), Qring.algebra_generator("y")
Q = x**2 + y**2
```

```python
s = Set([Q(u, v) for u in range(n) for v in range(n)])
l = sorted(s); l[0:20]
```

```python
L1 = [k % 4 for k in s]
L1[0:10]
```

```python
s2 = Set(range(2*n^2))
sorted(s2)[0:10]
```

```python
L2 = [k % 4 for k in s2.difference(s)]
L2[0:10]
```

```python
C1 = Counter(L1)
C1
```

```python
C2 = Counter(L2)
C2
```
