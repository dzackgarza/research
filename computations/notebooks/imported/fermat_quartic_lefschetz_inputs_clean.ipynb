```sage
import time
begin = time.time()
os.environ["SAGE_NUM_THREADS"] = '10'
from lefschetz_family import Hypersurface

import logging
logging.basicConfig()
logging.getLogger('lefschetz_family.hypersurface').setLevel(logging.INFO)
S.<X,Y,Z,W> =PolynomialRing(QQ)
```

We will compute the homology of this polynomial.


```sage
P = X^4+Y^4+Z^4+W^4
show(P)
```

```sage
fibration = [vector([0, 0, 0, 1]), vector([2, 3, 1, 0]), vector([7, -10, -4, -8])]
surface = Hypersurface(P, fibration=fibration, method="voronoi", nbits=1000)
```

We need to have an effective period matrix of the fiber. We care about the basis of cohomology in which this matrix is given, but not about the basis of homology (although we have it explicitely).


```sage
surface.fiber.cohomology
```

```sage
show(surface.fiber.period_matrix)
surface.fiber.period_matrix.dimensions()
```

For each critical point we compute a simple loop around it:


```sage
path = surface.paths[30]
```

```sage
(
    list_plot([[z.real(), z.imag()] for z in surface.critical_values], color="red") + 
    list_plot([[z.real(), z.imag()] for z in path], True, color="orange")
)
```

We additionally order the loops in a way that makes their composition the loop around $\infty$:


```sage
from lefschetz_family.util import Util

paths = surface.paths
pathtot = []
for p in paths:
    pathtot= pathtot+p
pathtot = Util.simplify_path(pathtot)
pathtot
```

```sage
(
    list_plot([[z.real(), z.imag()] for z in surface.critical_values], color="red") + 
    list_plot([[z.real(), z.imag()] for z in pathtot], True, color="orange")
)
```

Then for each of these loops $\ell$ we compute the matrix of the action of monodromy along $\ell$ in $H_{n-1}(Y_b)$ (given in the basis of homology of the period matrix).


```sage
surface.fiber.cohomology
```

The fibre is a quartic curve, it has genus $3$.


```sage
surface.fiber.homology
```

```sage
surface.family.pol
```

```sage
surface._restrict_form(surface.cohomology[0])
```

To compute the monodromy matrices, we integrate the Picard-Fuchs operator of the holomorphic form. It has order $6$ and degree $129$.


```sage
surface.picard_fuchs_equation(0).degree(), surface.picard_fuchs_equation(0).order()
```

```sage
surface.monodromy_matrices
```

From then on we can easily recover everything we need.


This is the transpose of the $36\times 6$ matrix of the boundary map $\delta : H_2(Y_+, Y_b)\to H_1(Y_b)$.


```sage
matrix(surface.vanishing_cycles).transpose()
```

The lines of the following $6\times 36$ matrix are the coefficients of the extensions around infinity in the basis of thimbles. They generate the image of the extension map $\tau^2_\infty:H_{n-1}(Y_b)\to H_{n}(Y_+)$. These extensions are trivial in $H_2(Y)$, but not in $H_2(Y_+)$. We check that they lie in $\ker\delta$.


```sage
matrix(surface.infinity_loops)
```

```sage
matrix(surface.infinity_loops)*matrix(surface.vanishing_cycles)
```

Cycles representing a basis of the $\frac{\ker\delta}{\operatorname{im}\tau_\infty^2}$ term in the description of $H_n(Y)$ in Theorem 1 can be obtained by finding a complement of $\operatorname{im}\tau_\infty^2$ in $\ker\delta$:


```sage
print(matrix(surface.extensions))
```

```sage
for i, c in enumerate(surface.extensions[1]):
    if c!=0:
        print(i+1," : ", c)
```
