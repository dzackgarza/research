```python
from collections import Counter
%matplotlib notebook

import matplotlib
import numpy as np
import matplotlib.pyplot as plt
n = 20
```

```python
s = Set()
Q(x, y) = x^2 + y^2
```

```python
s = Set( [ ZZ( Q(x, y) ) for x in sequence(0, n) for y in sequence(0, n) ] )
l = s.list(); l.sort(); l[0:20]
```

```python
L1 = [k % 4 for k in s.list() ]
L1[0:10]
```

```python
s2 = Set( sequence(0, 2*n^2) )
s2[0:10]
```

```python
L2 = [ k % 4 for k in s2.difference(s) ]
L2[0:10]
```

```python
C1 = Counter(L1)
C1
```

```python
C2 = Counter(L2)
C2
```

```python
plt.bar(C1.keys(), C1.values())
```
