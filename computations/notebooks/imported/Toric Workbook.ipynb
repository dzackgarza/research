```python
# Create a list of vectors in R^2 representing the generators of the toric variety
vectors = [(1, 0), (0, 1), (-1, -1)]

# Create a polyhedron from the vectors
polyhedron = Polyhedron(vertices=vectors)

# Create a toric variety from the polyhedron
variety = ToricVariety(polyhedron)
```

```python
# Compute the self-intersection numbers of the toric variety
intersection_numbers = variety.intersection_numbers()

# Print the self-intersection numbers
print(intersection_numbers)
```
