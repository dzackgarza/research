```python
def GS(a):
    u = [None] * len(a)
    e = [None] * len(a)
    for k in range(len(a)):
        projs = [P(u[j], a[k]) for j in range(k)]
        show(k, projs)
        u[k] = (a[k] - sum(projs))
        e[k] = (u[k]).normalized()
    return (u, e)
```

```python
A = Matrix(SR, [[12,-51,4], [6,167,-68],[-4,24,-41]])
show(A)
GS(A.columns())
print "done"
```
