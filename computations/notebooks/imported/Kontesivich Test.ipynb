```python
t1(d, k) = binomial(3*d-4, 3*k-2)
t2(d, k) = binomial(3*d-4, 3*k-1)

def N(d):
    if d == 1 or d == 2:
        return 1
    terms = [(i, d-i) for i in range(d)][1:d] 
    soln = 0
    for (k, l) in terms:
        soln += N(k) * N(l) * (k^2 * l^2 * t1(d, k) - k^3*l *t2(d, k))
    return(soln)
```

```python
[print(d, ": ", f"{int(N(d)):,}") for d in range(15)]
```
