## Mapping spaces $C^k(X,Y)$

$C^k(X,Y)$ is the space of maps $X\to Y$ of class $C^k$, a subset of $\operatorname{Hom}_{\mathbf{Set}}(X,Y)$. Regularity $k$ is a parameter; $\infty$ is a value of $k$. The shorthand $C^k(X)$ is $C^k(X,X)$.

When the codomain is an $\mathbb{R}$-module, $C^k(X,Y)$ is an object of $\mathrm{Vect}_{\mathbb{R}}$. When the values are $\mathbb{R}$, pointwise product makes $C^k(X,\mathbb{R})$ a commutative $\mathbb{R}$-algebra.

## Lebesgue classes

$L^p(\mathbb{R})$ is an $\mathbb{R}$-module of function representatives. Hölder pairs $L^p$ with $L^{p'}$ when $1/p+1/p'=1$: that pairing module is $L^p * L^{p'}$, equivalently `Lp(p).pairing_module()`, and evaluates by $X.\mathrm{pairing}(f,g)=\int fg$. On the diagonal $p=p'=2$, $L^2$ is a formed module, so it also has $b(f,g)$ and $q(f)$.

## Sequence spaces $\ell^p$

$\ell^p(\mathbb{R})$ is the $\mathbb{R}$-module of $p$-summable sequences $\mathbb{N}\to\mathbb{R}$. Hölder pairs $\ell^p$ with $\ell^{p'}$ when $1/p+1/p'=1$: that pairing module is $\ell^p * \ell^{p'}$, equivalently `ell(p).pairing_module()`, and evaluates by $X.\mathrm{pairing}(a,c)=\sum a_n c_n$. On the diagonal $p=p'=2$, $\ell^2$ is a formed module, so it also has $b(a,c)$ and $q(a)$.


```python
ell^2(RR)
```
