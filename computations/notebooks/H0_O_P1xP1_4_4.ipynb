# Generators of $H^0(\mathbf P^1\times\mathbf P^1,\mathcal O(a,b))$

Let $[x_0:x_1]$ and $[y_0:y_1]$ be homogeneous coordinates on the two factors.
By the Künneth isomorphism,

$$
H^0(\mathbf P^1\times\mathbf P^1,\mathcal O(a,b))
\cong H^0(\mathbf P^1,\mathcal O(a))\otimes H^0(\mathbf P^1,\mathcal O(b))
\cong \operatorname{Sym}^a(k^2)^\vee\otimes\operatorname{Sym}^b(k^2)^\vee ,
$$

so a basis consists of the bihomogeneous monomials

$$
x_0^{a-i}x_1^i\,y_0^{b-j}y_1^j,\qquad 0\le i\le a,\quad 0\le j\le b,
$$

and the dimension is $(a+1)(b+1)$. Everything below is parameterized in $(a,b)$;
the case of interest is $(4,4)$, where the dimension is $25$.

```python
from sage.version import version as sage_version
show(sage_version)
```

```python
a, b = 4, 4

R.<x0, x1, y0, y1> = PolynomialRing(QQ)
show(R)
```

## Monomial basis

```python
basis = [
    x0^(a - i) * x1^i * y0^(b - j) * y1^j
    for i in range(a + 1)
    for j in range(b + 1)
]
show(basis)
```

```python
def bidegree(m):
    e0, e1, f0, f1 = m.exponents()[0]
    return (e0 + e1, f0 + f1)


assert len(basis) == (a + 1) * (b + 1) == 25
assert len(set(basis)) == len(basis)
assert all(bidegree(m) == (a, b) for m in basis)

show(LatexExpr(r"\dim H^0(\mathbf P^1\times\mathbf P^1,\mathcal O(4,4))=25"))
```

## Generic section

A general section is the basis expansion with indeterminate coefficients,

$$
F=\sum_{i=0}^{a}\sum_{j=0}^{b} c_{ij}\,x_0^{a-i}x_1^i\,y_0^{b-j}y_1^j ,
$$

so $F$ lives in $S=A[x_0,x_1,y_0,y_1]$ over the coefficient ring
$A=k[c_{ij}]$, whose $(a+1)(b+1)$ variables are the projective coordinates on
$\mathbf P H^0(\mathcal O(a,b))$.

```python
A = PolynomialRing(QQ, [f"c_{i}_{j}" for i in range(a + 1) for j in range(b + 1)])
S = PolynomialRing(A, names=("x0", "x1", "y0", "y1"))
X0, X1, Y0, Y1 = S.gens()
c = A.gens()

F = sum(
    c[(b + 1) * i + j] * X0^(a - i) * X1^i * Y0^(b - j) * Y1^j
    for i in range(a + 1)
    for j in range(b + 1)
)
show(F)
```

```python
assert A.ngens() == (a + 1) * (b + 1)
assert set(F.monomials()) == {S(m) for m in basis}
assert sorted(F.coefficients()) == sorted(c)

show(LatexExpr(r"F \text{ is the generic section: its monomials are exactly the basis above}"))
```

## $\tau$-invariant monomials

We let $\tau$ act by

$$
\tau([x_0:x_1],[y_0:y_1])=([x_0:-x_1],[y_0:-y_1]),
$$

which restricts on the affine chart $x_0y_0\neq 0$ to $(x,y)\mapsto(-x,-y)$. We compute the induced action on the monomial basis and extract the fixed monomials.

```python
def tau_on_monomial(m):
    return m(x0=x0, x1=-x1, y0=y0, y1=-y1)


tau_images = [tau_on_monomial(m) for m in basis]
tau_invariant_monomials = [m for m, tm in zip(basis, tau_images) if tm == m]
tau_anti_invariant_monomials = [m for m, tm in zip(basis, tau_images) if tm == -m]

assert all(tm in (m, -m) for m, tm in zip(basis, tau_images))
assert len(tau_invariant_monomials) + len(tau_anti_invariant_monomials) == len(basis)
assert all(tau_on_monomial(m) == m for m in tau_invariant_monomials)
assert all(tau_on_monomial(m) == -m for m in tau_anti_invariant_monomials)

show(tau_invariant_monomials)
show(LatexExpr(r"\dim H^0(\mathcal O(4,4))^{\tau}=" + str(len(tau_invariant_monomials))))
```

## Reusable projective-scheme operations

Let $k$ be the base field. The notebook uses the following scheme-theoretic constructions for projective spaces, products of projective spaces, and supported closed subschemes:

- the categorical product $X\times_k Y$;
- the diagonal morphism $\Delta_X=(\operatorname{id}_X,\operatorname{id}_X):X\to X\times_k X$;
- the scheme-theoretic image of a morphism;
- the inverse image $f^{-1}(Z)=X\times_Y Z$ of a closed subscheme $Z\subseteq Y$;
- the graph morphism $\gamma_f=(\operatorname{id}_X,f):X\to X\times_k Y$;
- the equalizer of two endomorphisms, specialized below to the fixed subscheme of one endomorphism.

The diagonal criterion for separatedness identifies a separated $k$-scheme with a closed subscheme of its square via $\Delta_X$; projective $k$-schemes are separated. See Hartshorne, *Algebraic Geometry*, Chapter II, Section 4. The next cell supplies notebook-local implementations of the operations not presently available for Sage products of projective spaces. Its elimination backend is restricted by explicit checks to the supported projective ambient spaces and a common base field.

```python
from sage.schemes.projective.projective_space import ProjectiveSpace_ring
from sage.schemes.product_projective.space import ProductProjectiveSpaces_ring
import sage.schemes.projective.projective_morphism as _projective_morphism_module
from sage.schemes.product_projective.morphism import ProductProjectiveSpaces_morphism_ring


def _supported_projective_ambient(space):
    ambient = space.ambient_space()
    return isinstance(
        ambient,
        (ProjectiveSpace_ring, ProductProjectiveSpaces_ring),
    )


def _require_supported_projective_ambient(space, operation):
    if not _supported_projective_ambient(space):
        raise NotImplementedError(
            f"{operation} is currently implemented only for projective spaces, "
            "products of projective spaces, and their closed subschemes"
        )
    return space.ambient_space()


def _factor_dimensions(space):
    ambient = _require_supported_projective_ambient(
        space,
        'factor decomposition',
    )
    if isinstance(ambient, ProjectiveSpace_ring):
        return (ZZ(ambient.dimension_relative()),)
    return tuple(
        ZZ(d) for d in ambient.dimension_relative_components()
    )


def _factor_blocks(space, coordinates):
    coordinates = tuple(coordinates)
    dimensions = _factor_dimensions(space)
    expected = sum(d + 1 for d in dimensions)
    if len(coordinates) != expected:
        raise ValueError(
            f"expected {expected} coordinates, received {len(coordinates)}"
        )

    blocks = []
    start = 0
    for dimension in dimensions:
        stop = start + dimension + 1
        blocks.append(coordinates[start:stop])
        start = stop
    return tuple(blocks)


def _defining_equations(space):
    ambient = space.ambient_space()
    if space == ambient:
        return tuple()
    if not hasattr(space, 'defining_polynomials'):
        raise NotImplementedError(
            'closed-subscheme equations are unavailable for this object'
        )
    return tuple(space.defining_polynomials())


def _saturate_projectively(ideal, coordinate_blocks):
    saturated = ideal
    for block in coordinate_blocks:
        saturated = saturated.saturation(
            ideal.ring().ideal(block)
        )[0]
    return saturated


def _scheme_theoretic_image_projective(self):
    source = self.domain()
    target = self.codomain()
    source_ambient = _require_supported_projective_ambient(
        source,
        'scheme-theoretic image',
    )
    target_ambient = _require_supported_projective_ambient(
        target,
        'scheme-theoretic image',
    )

    base = source.base_ring()
    if base != target.base_ring() or not base.is_field():
        raise NotImplementedError(
            'the current elimination backend requires a common base field'
        )

    source_dimensions = _factor_dimensions(source)
    target_dimensions = _factor_dimensions(target)
    source_count = sum(d + 1 for d in source_dimensions)
    target_count = sum(d + 1 for d in target_dimensions)
    image_coordinates = tuple(self.defining_polynomials())
    if len(image_coordinates) != target_count:
        raise ValueError(
            'the morphism coordinate count does not match the target factors'
        )

    names = (
        tuple(f'src{i}' for i in range(source_count))
        + tuple(f'tgt{i}' for i in range(target_count))
    )
    joint_ring = PolynomialRing(
        base,
        names=names,
        order='lex',
    )
    joint_coordinates = joint_ring.gens()
    source_coordinates = joint_coordinates[:source_count]
    target_coordinates = joint_coordinates[source_count:]

    source_ring = source_ambient.coordinate_ring()
    target_ring = target_ambient.coordinate_ring()
    source_into_joint = source_ring.hom(
        source_coordinates,
        joint_ring,
    )
    target_into_joint = target_ring.hom(
        target_coordinates,
        joint_ring,
    )

    graph_generators = [
        source_into_joint(f)
        for f in _defining_equations(source)
    ]
    graph_generators.extend(
        target_into_joint(f)
        for f in _defining_equations(target)
    )

    image_coordinates_joint = tuple(
        source_into_joint(f)
        for f in image_coordinates
    )
    for target_block, image_block in zip(
        _factor_blocks(target, target_coordinates),
        _factor_blocks(target, image_coordinates_joint),
    ):
        comparison = matrix(
            joint_ring,
            [target_block, image_block],
        )
        graph_generators.extend(comparison.minors(2))

    graph_ideal = joint_ring.ideal(graph_generators)
    graph_ideal = _saturate_projectively(
        graph_ideal,
        _factor_blocks(source, source_coordinates),
    )
    eliminated = graph_ideal.elimination_ideal(
        source_coordinates
    )

    joint_to_target = joint_ring.hom(
        [target_ring(0)] * source_count
        + list(target_ring.gens()),
        target_ring,
    )
    image_ideal = target_ring.ideal([
        joint_to_target(f)
        for f in eliminated.gens()
    ])
    image_ideal = _saturate_projectively(
        image_ideal,
        _factor_blocks(target, target_ring.gens()),
    )
    return target_ambient.subscheme(image_ideal.gens())


def _image_dispatch(self):
    if (
        _supported_projective_ambient(self.domain())
        and _supported_projective_ambient(self.codomain())
    ):
        return _scheme_theoretic_image_projective(self)

    original = getattr(
        type(self),
        '_notebook_original_image',
        None,
    )
    if original is None:
        raise NotImplementedError(
            'no image backend is available for this morphism'
        )
    return original(self)


def _diagonal_morphism(self):
    _require_supported_projective_ambient(
        self,
        'diagonal morphism',
    )
    if self != self.ambient_space():
        raise NotImplementedError(
            'diagonal morphisms of closed subschemes are not yet implemented'
        )

    square = self * self
    coordinates = tuple(self.gens())
    return self.hom(
        coordinates + coordinates,
        square,
    )


def _diagonal_subscheme(self):
    return self.diagonal_morphism().image()


def _graph_morphism(self):
    if self.domain() != self.codomain():
        raise ValueError(
            'graph_morphism requires an endomorphism'
        )

    X = self.domain()
    _require_supported_projective_ambient(
        X,
        'graph morphism',
    )
    if X != X.ambient_space():
        raise NotImplementedError(
            'graphs from closed subschemes are not yet implemented'
        )

    target = X * X
    identity_coordinates = tuple(
        X.identity_morphism().defining_polynomials()
    )
    image_coordinates = tuple(
        self.defining_polynomials()
    )
    return X.hom(
        identity_coordinates + image_coordinates,
        target,
    )


def _inverse_image_subscheme(self, subscheme):
    source = self.domain()
    target = self.codomain()
    source_ambient = _require_supported_projective_ambient(
        source,
        'inverse image',
    )
    target_ambient = _require_supported_projective_ambient(
        target,
        'inverse image',
    )

    if subscheme.ambient_space() != target_ambient:
        raise ValueError(
            'the subscheme must lie in the ambient space of the codomain'
        )

    source_ring = source_ambient.coordinate_ring()
    target_ring = target_ambient.coordinate_ring()
    pullback_map = target_ring.hom(
        tuple(self.defining_polynomials()),
        source_ring,
    )

    equations = list(_defining_equations(source))
    equations.extend(
        pullback_map(f)
        for f in subscheme.defining_polynomials()
    )
    inverse_image_ideal = source_ring.ideal(equations)
    inverse_image_ideal = _saturate_projectively(
        inverse_image_ideal,
        _factor_blocks(source, source_ring.gens()),
    )
    return source_ambient.subscheme(
        inverse_image_ideal.gens()
    )


def _fixed_subscheme(self):
    if self.domain() != self.codomain():
        raise ValueError(
            'fixed_subscheme requires an endomorphism'
        )

    X = self.domain()
    diagonal = X.diagonal_morphism().image()
    graph = self.graph_morphism()
    return graph.inverse_image_subscheme(diagonal)


_projective_morphism_classes = (
    _projective_morphism_module.SchemeMorphism_polynomial_projective_space,
    _projective_morphism_module.SchemeMorphism_polynomial_projective_space_field,
    _projective_morphism_module.SchemeMorphism_polynomial_projective_space_finite_field,
    ProductProjectiveSpaces_morphism_ring,
)

for morphism_class in _projective_morphism_classes:
    if not hasattr(
        morphism_class,
        '_notebook_original_image',
    ):
        morphism_class._notebook_original_image = getattr(
            morphism_class,
            'image',
            None,
        )

    morphism_class.scheme_theoretic_image = (
        _scheme_theoretic_image_projective
    )
    morphism_class.image = _image_dispatch
    morphism_class.inverse_image_subscheme = (
        _inverse_image_subscheme
    )
    morphism_class.pullback_subscheme = (
        _inverse_image_subscheme
    )
    morphism_class.graph_morphism = _graph_morphism
    morphism_class.fixed_subscheme = _fixed_subscheme

for space_class in (
    ProjectiveSpace_ring,
    ProductProjectiveSpaces_ring,
):
    space_class.diagonal_morphism = _diagonal_morphism
    space_class.diagonal_subscheme = _diagonal_subscheme


# Regression tests for products, images, pullbacks, and higher-order maps.
P1_test_x = ProjectiveSpace(
    QQ,
    1,
    names=('tx0', 'tx1'),
)
P1_test_y = ProjectiveSpace(
    QQ,
    1,
    names=('ty0', 'ty1'),
)
X_test = P1_test_x * P1_test_y
assert X_test.n_components() == 2
assert (X_test * X_test).n_components() == 4

Delta_test = X_test.diagonal_morphism()
Diagonal_test = Delta_test.image()
assert Diagonal_test.dimension() == 2

P1_test_a = ProjectiveSpace(
    QQ,
    1,
    names=('ta0', 'ta1'),
)
P1_test_b = ProjectiveSpace(
    QQ,
    1,
    names=('tb0', 'tb1'),
)
P1_test_c = ProjectiveSpace(
    QQ,
    1,
    names=('tc0', 'tc1'),
)
T_test = P1_test_a * P1_test_b * P1_test_c
ta0, ta1, tb0, tb1, tc0, tc1 = T_test.gens()
sigma_test = T_test.hom(
    [tb0, tb1, tc0, tc1, ta0, ta1],
    T_test,
)
Fix_sigma_test = sigma_test.fixed_subscheme()
assert Fix_sigma_test.dimension() == 1

print(
    'Projective product/image/pullback framework loaded; '
    'all regression tests passed.'
)
```

## Products and the diagonal

The product $X\times_k Y$ is characterized by the natural bijection

$$
\operatorname{Hom}_k(T,X\times_k Y)
\cong
\operatorname{Hom}_k(T,X)\times\operatorname{Hom}_k(T,Y)
$$

for every $k$-scheme $T$. Taking both components to be $\operatorname{id}_X$ gives the diagonal morphism

$$
\Delta_X:X\longrightarrow X\times_k X,
\qquad
x\longmapsto(x,x).
$$

Equivalently, if $p_1,p_2:X\times_k X\to X$ are the projections, then $\Delta_X$ is the unique morphism satisfying

$$
p_1\circ\Delta_X=\operatorname{id}_X,
\qquad
p_2\circ\Delta_X=\operatorname{id}_X.
$$

We construct two copies of $\mathbf P^1$ independently and use Sage's native product operation. Consequently the square is formed as $X*X$ rather than by manually listing four projective factors.

```python
P1_x = ProjectiveSpace(
    QQ,
    1,
    names=('x0', 'x1'),
)
P1_y = ProjectiveSpace(
    QQ,
    1,
    names=('y0', 'y1'),
)

X = P1_x * P1_y
X_square = X * X
Delta_X = X.diagonal_morphism()

show(P1_x)
show(P1_y)
show(X)
show(X_square)
show(Delta_X)
```

## Scheme-theoretic image of the diagonal

For a morphism $f:S\to T$, its scheme-theoretic image is the smallest closed subscheme $Z\subseteq T$ through which $f$ factors. In sheaf-theoretic terms, its ideal sheaf is

$$
\ker\!\left(\mathcal O_T\longrightarrow f_*\mathcal O_S\right)
$$

whenever this description is formed in the usual quasi-coherent setting. For the diagonal of a separated scheme, the morphism $\Delta_X$ is already a closed immersion, so its scheme-theoretic image is the diagonal closed subscheme of $X\times_k X$.

The computational backend realizes this construction in multihomogeneous coordinates. It forms the graph ideal in independent source and target variables, saturates by the irrelevant ideals of the projective source factors, and eliminates the source variables. The resulting saturated target ideal defines the scheme-theoretic image. Thus the equations displayed below are derived from the morphism $\Delta_X$ by elimination; they are not supplied as diagonal equations by hand.

```python
Diagonal_X = Delta_X.image()

show(Diagonal_X)
show(Diagonal_X.defining_polynomials())
```

## Coordinate form of the diagonal

Write the coordinates on the two copies of $X$ in $X\times_k X$ as

$$
([x_0^L:x_1^L],[y_0^L:y_1^L])
\quad\text{and}\quad
([x_0^R:x_1^R],[y_0^R:y_1^R]).
$$

Equality in each projective factor is detected by rank one of the matrix whose rows are the two homogeneous coordinate vectors. We therefore compare the ideal obtained from `Delta_X.image()` with the ideal generated by the $2\times2$ minors of these factorwise comparison matrices. The code extracts the factor blocks from the product structure of $X\times X$; no coordinate equation of the diagonal is entered manually.

```python
square_blocks = _factor_blocks(
    X_square,
    X_square.gens(),
)
n_X_factors = X.n_components()
left_blocks = square_blocks[:n_X_factors]
right_blocks = square_blocks[n_X_factors:]

comparison_matrices_diagonal = [
    matrix(
        X_square.coordinate_ring(),
        [left_block, right_block],
    )
    for left_block, right_block in zip(
        left_blocks,
        right_blocks,
    )
]
diagonal_minors = tuple(
    minor
    for comparison_matrix in comparison_matrices_diagonal
    for minor in comparison_matrix.minors(2)
)
diagonal_image_equations = tuple(
    Diagonal_X.defining_polynomials()
)

I_diagonal_image = X_square.coordinate_ring().ideal(
    diagonal_image_equations
)
I_diagonal_minors = X_square.coordinate_ring().ideal(
    diagonal_minors
)
assert I_diagonal_image == I_diagonal_minors

for i, comparison_matrix in enumerate(
    comparison_matrices_diagonal,
    start=1,
):
    show(LatexExpr(r"C_" + str(i) + "="), comparison_matrix)
    show(
        LatexExpr(r"\det(C_" + str(i) + ")="),
        comparison_matrix.det(),
    )

show(LatexExpr(r"I_{\Delta_X}="), I_diagonal_image)
```

## Graphs, pullbacks, and the fixed subscheme

Let $\tau:X\to X$ be an endomorphism. Its graph morphism is

$$
\gamma_\tau=(\operatorname{id}_X,\tau):X\longrightarrow X\times_k X.
$$

For a morphism $f:S\to T$ and a closed subscheme $Z\subseteq T$ with ideal sheaf $\mathcal I_Z$, the scheme-theoretic inverse image $f^{-1}(Z)=S\times_T Z$ is defined by the image of

$$
f^{-1}\mathcal I_Z\otimes_{f^{-1}\mathcal O_T}\mathcal O_S
\longrightarrow
\mathcal O_S.
$$

In homogeneous coordinates this becomes pullback of the defining equations along $f$, followed by projective saturation. This is the operation implemented by `pullback_subscheme()`.

The fixed-point functor assigns to a $k$-scheme $T$ the set of morphisms $u:T\to X$ satisfying

$$
\tau\circ u=u.
$$

By the defining property of the diagonal, this condition is equivalent to requiring that

$$
(u,\tau\circ u):T\longrightarrow X\times_k X
$$

factor through $\Delta_X$. Hence the fixed-point functor is represented by the fiber product

$$
\operatorname{Fix}(\tau)
=
X\times_{X\times_k X}\Delta_X,
$$

where the map from the first copy of $X$ is $\gamma_\tau$. Equivalently,

$$
\operatorname{Fix}(\tau)
=
\gamma_\tau^{-1}(\Delta_X).
$$

This is an equalizer construction, not merely a search for fixed rational points. In particular, the inverse image retains the scheme structure, including possible nilpotents or multiplicities. The order of $\tau$ does not enter the construction; the same operation applies to any endomorphism, and replacing $\tau$ by $\tau^m$ computes the fixed subscheme of that iterate.

```python
x0, x1, y0, y1 = X.gens()
tau = X.hom(
    [x0, -x1, y0, -y1],
    X,
)
f0, f1, g0, g1 = tau.defining_polynomials()
graph_tau = tau.graph_morphism()

show(tau)
show(LatexExpr(r"f_0="), f0)
show(LatexExpr(r"f_1="), f1)
show(LatexExpr(r"g_0="), g0)
show(LatexExpr(r"g_1="), g1)
show(graph_tau)
```

## Coordinate specialization of the pullback

The graph morphism induces a homomorphism on multihomogeneous coordinate rings

$$
\gamma_\tau^*:k[X\times_k X]\longrightarrow k[X].
$$

On the left copy of $X$, this homomorphism substitutes the source coordinates; on the right copy, it substitutes the defining coordinates $(f_0,f_1,g_0,g_1)$ of $\tau$. Therefore the ideal of the inverse image of the diagonal is obtained by applying $\gamma_\tau^*$ to the equations already computed for $\Delta_X(X)$ and then saturating projectively.

To compare this abstract pullback with the earlier coordinate calculation, we also form the factorwise source/image comparison matrices from the defining polynomials of $\tau$. The code verifies equality of the resulting ideals rather than assuming the determinant equations.

```python
graph_coordinates = tuple(
    graph_tau.defining_polynomials()
)
graph_coordinate_pullback = X_square.coordinate_ring().hom(
    graph_coordinates,
    X.coordinate_ring(),
)
pulled_back_diagonal_equations = tuple(
    graph_coordinate_pullback(equation)
    for equation in diagonal_image_equations
)

source_blocks_tau = _factor_blocks(
    X,
    X.identity_morphism().defining_polynomials(),
)
image_blocks_tau = _factor_blocks(
    X,
    tau.defining_polynomials(),
)
comparison_matrices_tau = [
    matrix(
        X.coordinate_ring(),
        [source_block, image_block],
    )
    for source_block, image_block in zip(
        source_blocks_tau,
        image_blocks_tau,
    )
]
fixed_minors = tuple(
    minor
    for comparison_matrix in comparison_matrices_tau
    for minor in comparison_matrix.minors(2)
)

I_pulled_back_diagonal = X.coordinate_ring().ideal(
    pulled_back_diagonal_equations
)
I_fixed_minors = X.coordinate_ring().ideal(
    fixed_minors
)
assert I_pulled_back_diagonal == I_fixed_minors

I_pulled_back_diagonal_saturated = _saturate_projectively(
    I_pulled_back_diagonal,
    _factor_blocks(X, X.coordinate_ring().gens()),
)
Fix_tau_via_pullback = graph_tau.pullback_subscheme(
    Diagonal_X
)
assert (
    I_pulled_back_diagonal_saturated
    == Fix_tau_via_pullback.defining_ideal()
)

for diagonal_equation, pulled_back_equation in zip(
    diagonal_image_equations,
    pulled_back_diagonal_equations,
):
    show(
        diagonal_equation,
        LatexExpr(r"\stackrel{\gamma_\tau^*}{\longmapsto}"),
        pulled_back_equation,
    )

for i, comparison_matrix in enumerate(
    comparison_matrices_tau,
    start=1,
):
    show(LatexExpr(r"M_" + str(i) + "="), comparison_matrix)
    show(
        LatexExpr(r"\det(M_" + str(i) + ")="),
        comparison_matrix.det(),
    )

show(
    LatexExpr(r"I_{\gamma_\tau^{-1}(\Delta_X)}="),
    I_pulled_back_diagonal_saturated,
)
```

## Fixed subscheme and rational fixed points

We now construct the fixed subscheme through the reusable equalizer method and compare it with the explicit pullback computed above. We then compute its dimension and its $\mathbf Q$-rational points and verify the fixed-point condition by evaluating the Sage morphism on each point.

```python
Fix_tau = tau.fixed_subscheme()
assert Fix_tau == Fix_tau_via_pullback

fixed_dimension = Fix_tau.dimension()
fixed_points = [
    X(P) for P in Fix_tau.rational_points()
]

assert fixed_dimension == 0
assert len(fixed_points) == 4
assert all(tau(P) == P for P in fixed_points)

show(Fix_tau)
show(
    LatexExpr(r"\dim \operatorname{Fix}(\tau)="),
    fixed_dimension,
)
print("Computed fixed points:")
for P in fixed_points:
    print(P)
show(
    LatexExpr(r"\#\operatorname{Fix}(\tau)(\mathbf Q)="),
    len(fixed_points),
)
```

## The family of branch curves under study

Let

$$
Y=\mathbf P^1\times\mathbf P^1,
\qquad
L=\mathcal O_Y(4,4)=-2K_Y.
$$

A nonzero section $s\in H^0(Y,L)$ determines the divisor $B_s=V(s)$, and scalar multiples determine the same divisor. Thus branch divisors are parametrized by the projective space $\mathbf P(H^0(Y,L))$.

We study the locus

$$
\mathcal A=
\left\{
[s]\in\mathbf P(H^0(Y,L)):
\tau(B_s)=B_s,
\ B_s\cap\operatorname{Fix}(\tau)=\varnothing,
\ B_s\text{ has at worst ADE singularities}
\right\}.
$$

We use the elementary linear-algebra fact that, for an involution acting linearly over a field of characteristic different from $2$, an invariant line in $\mathbf P(V)$ is represented by an eigenvector. We therefore compute the two eigenspaces of $\tau^*$ on $H^0(Y,L)$, test their evaluation at the fixed locus, and then study the smooth locus inside the surviving projective eigenspace.

```python
S44 = PolynomialRing(QQ, names=('X0', 'X1', 'Y0', 'Y1'))
X0, X1, Y0, Y1 = S44.gens()

bidegree_44_basis = [
    X0**(4-i) * X1**i * Y0**(4-j) * Y1**j
    for i in range(5)
    for j in range(5)
]


def tau_on_section(section):
    return section(X0=X0, X1=-X1, Y0=Y0, Y1=-Y1)


invariant_basis = [
    monomial
    for monomial in bidegree_44_basis
    if tau_on_section(monomial) == monomial
]
anti_invariant_basis = [
    monomial
    for monomial in bidegree_44_basis
    if tau_on_section(monomial) == -monomial
]

V_dimension = len(bidegree_44_basis)
V_plus_dimension = len(invariant_basis)
V_minus_dimension = len(anti_invariant_basis)
P_V_plus_dimension = V_plus_dimension - 1
P_V_minus_dimension = V_minus_dimension - 1

assert V_plus_dimension + V_minus_dimension == V_dimension
assert all(tau_on_section(m) == m for m in invariant_basis)
assert all(tau_on_section(m) == -m for m in anti_invariant_basis)

show(LatexExpr(r'\dim H^0(Y,L)='), V_dimension)
show(LatexExpr(r'\dim V_+='), V_plus_dimension)
show(LatexExpr(r'\dim V_-='), V_minus_dimension)
show(LatexExpr(r'\dim \mathbf P(V_+)='), P_V_plus_dimension)
show(LatexExpr(r'\dim \mathbf P(V_-)='), P_V_minus_dimension)
show(LatexExpr(r'\text{basis}(V_+)='), invariant_basis)
show(LatexExpr(r'\text{basis}(V_-)='), anti_invariant_basis)
```

## Evaluation at the four fixed points

Let

$$
E_\pm:V_\pm\longrightarrow
\bigoplus_{p\in\operatorname{Fix}(\tau)}L|_p
$$

be evaluation at the fixed locus, using the coordinate representatives computed above. The direct sum has no preferred ordering of its four summands, so the computation below makes no order-sensitive assertion. Instead it records:

- the set of fixed points;
- the rank and kernel dimension of evaluation;
- the set of invariant monomials detected by evaluation;
- the ideal generated by the four generic evaluation values;
- the principal ideal defining the complement of the fixed-point-avoidance locus.

This distinguishes the kernel of evaluation, whose divisors contain fixed points, from the principal open locus on which all four evaluations are nonzero.

```python
exponent_pairs = [
    (i, j)
    for i in range(5)
    for j in range(5)
]
monomial_by_pair = {
    (i, j): X0**(4-i) * X1**i * Y0**(4-j) * Y1**j
    for i, j in exponent_pairs
}
invariant_pairs = tuple(
    pair
    for pair in exponent_pairs
    if monomial_by_pair[pair] in invariant_basis
)
anti_invariant_pairs = tuple(
    pair
    for pair in exponent_pairs
    if monomial_by_pair[pair] in anti_invariant_basis
)

corner_pair_set = frozenset({
    (0, 0),
    (0, 4),
    (4, 0),
    (4, 4),
})
expected_fixed_point_set = frozenset({
    ((QQ(1), QQ(0)), (QQ(1), QQ(0))),
    ((QQ(1), QQ(0)), (QQ(0), QQ(1))),
    ((QQ(0), QQ(1)), (QQ(1), QQ(0))),
    ((QQ(0), QQ(1)), (QQ(0), QQ(1))),
})
computed_fixed_point_set = frozenset(
    (tuple(P[0]), tuple(P[1]))
    for P in fixed_points
)
assert computed_fixed_point_set == expected_fixed_point_set

# Any row order may be used; all assertions below are invariant under permutation.
fixed_coordinate_points = tuple(computed_fixed_point_set)


def evaluate_section_at_fixed_point(section, point):
    (a0, a1), (b0, b1) = point
    return QQ(section(X0=a0, X1=a1, Y0=b0, Y1=b1))


evaluation_matrix_plus = matrix(QQ, [
    [
        evaluate_section_at_fixed_point(monomial, point)
        for monomial in invariant_basis
    ]
    for point in fixed_coordinate_points
])
evaluation_matrix_minus = matrix(QQ, [
    [
        evaluate_section_at_fixed_point(monomial, point)
        for monomial in anti_invariant_basis
    ]
    for point in fixed_coordinate_points
])

coefficient_names = [
    f'a{i}{j}' for i, j in invariant_pairs
]
CoefficientRing = PolynomialRing(QQ, names=coefficient_names)
invariant_coefficients = CoefficientRing.gens()
coefficient_by_pair = dict(
    zip(invariant_pairs, invariant_coefficients)
)

generic_fixed_values = tuple(
    sum(
        invariant_coefficients[k] * evaluation_matrix_plus[row, k]
        for k in range(V_plus_dimension)
    )
    for row in range(evaluation_matrix_plus.nrows())
)

row_supports_plus = tuple(
    tuple(
        column
        for column in range(evaluation_matrix_plus.ncols())
        if evaluation_matrix_plus[row, column] != 0
    )
    for row in range(evaluation_matrix_plus.nrows())
)
detected_pair_set = frozenset(
    invariant_pairs[support[0]]
    for support in row_supports_plus
)
corner_coefficients = tuple(
    coefficient_by_pair[pair]
    for pair in invariant_pairs
    if pair in corner_pair_set
)

evaluation_value_ideal = CoefficientRing.ideal(generic_fixed_values)
corner_coefficient_ideal = CoefficientRing.ideal(corner_coefficients)
evaluation_product_ideal = CoefficientRing.ideal(
    prod(generic_fixed_values)
)
corner_product_ideal = CoefficientRing.ideal(
    prod(corner_coefficients)
)

E_plus_rank = evaluation_matrix_plus.rank()
E_plus_kernel_dimension = evaluation_matrix_plus.right_kernel().dimension()
E_minus_is_zero = evaluation_matrix_minus.is_zero()

assert E_plus_rank == 4
assert E_plus_kernel_dimension == 9
assert E_minus_is_zero
assert all(len(support) == 1 for support in row_supports_plus)
assert detected_pair_set == corner_pair_set
assert evaluation_value_ideal == corner_coefficient_ideal
assert evaluation_product_ideal == corner_product_ideal

print('Computed fixed-point set:')
for point in computed_fixed_point_set:
    print(point)
print('Evaluation rows (point, detected monomial pair, generic value):')
for row, point in enumerate(fixed_coordinate_points):
    pair = invariant_pairs[row_supports_plus[row][0]]
    print(point, pair, generic_fixed_values[row])
show(LatexExpr(r'\operatorname{rank}(E_+)='), E_plus_rank)
show(LatexExpr(r'\dim\ker(E_+)='), E_plus_kernel_dimension)
show(LatexExpr(r'E_-=0\;\text{is}'), E_minus_is_zero)
show(LatexExpr(r'\text{detected monomial pairs}='), sorted(detected_pair_set))
show(LatexExpr(r'(\operatorname{im}E_+)^*\text{-coordinate ideal}='), evaluation_value_ideal)
```

## The fixed-point-avoidance locus

Inside $\mathbf P(V_+)$, define the avoidance locus by the nonvanishing of all four evaluated coordinates. The following cells compute its principal-open equation and its dimension.

The same evaluation computation tests whether $\mathbf P(V_-)$ contributes: if the evaluation map on $V_-$ is zero, every anti-invariant divisor contains the fixed locus and is excluded from $\mathcal A$.

```python
evaluation_avoidance_polynomial = prod(generic_fixed_values)
avoidance_polynomial = prod(corner_coefficients)
avoidance_principal_ideal = CoefficientRing.ideal(
    avoidance_polynomial
)
anti_invariant_avoidance_possible = not E_minus_is_zero
avoidance_locus_dimension = P_V_plus_dimension

assert (
    CoefficientRing.ideal(evaluation_avoidance_polynomial)
    == avoidance_principal_ideal
)
assert not anti_invariant_avoidance_possible
```

```python
show(
    LatexExpr(
        r'\mathcal U_{\mathrm{avoid}}='
        + r'D_+\!\left('
        + latex(avoidance_polynomial)
        + r'\right)\subset\mathbf P(V_+)'
    )
)
show(
    LatexExpr(r'\dim\mathcal U_{\mathrm{avoid}}='),
    avoidance_locus_dimension,
)
print(
    'Computed fixed-point condition: '
    + str(avoidance_polynomial)
    + ' != 0.'
)
print(
    'Anti-invariant eigenspace contributes an avoiding divisor:',
    anti_invariant_avoidance_possible,
)
```

## The ADE condition via a smooth open sublocus

Let

$$
\mathcal U_{\mathrm{smooth,avoid}}
=
\mathcal U_{\mathrm{smooth}}\cap\mathcal U_{\mathrm{avoid}}
\subset \mathbf P(V_+).
$$

We do not compute equations for the full at-worst-ADE locus. Instead we use the inclusions

$$
\mathcal U_{\mathrm{smooth,avoid}}
\subseteq
\mathcal A
\subseteq
\mathcal U_{\mathrm{avoid}}
\subseteq
\mathbf P(V_+),
$$

where $\mathcal A$ denotes the invariant, fixed-point-avoiding, at-worst-ADE locus. Smooth divisors are included because smoothness is stronger than having at worst ADE singularities.

We compute that the invariant linear system is base-point-free, invoke Bertini's theorem in characteristic zero to obtain a dense open smooth locus, and verify an explicit smooth member lying in the avoidance locus. The code then computes the resulting lower and upper bounds for $\dim\mathcal A$.

```python
base_ideal = S44.ideal(invariant_basis)
saturated_base_ideal = _saturate_projectively(
    base_ideal,
    ((X0, X1), (Y0, Y1)),
)
base_locus_is_empty = saturated_base_ideal.is_one()

smooth_example = (
    X0**4 * Y0**4
    + X0**4 * Y0**2 * Y1**2
    + X0**4 * Y1**4
    + X1**4 * Y0**4
    + X1**4 * Y1**4
)


def singular_ideal_on_chart(section, x_chart, y_chart):
    AffineChartRing = PolynomialRing(QQ, names=('u', 'v'))
    u, v = AffineChartRing.gens()
    x_substitution = (QQ(1), u) if x_chart == 0 else (u, QQ(1))
    y_substitution = (QQ(1), v) if y_chart == 0 else (v, QQ(1))
    chart_equation = AffineChartRing(section(
        X0=x_substitution[0],
        X1=x_substitution[1],
        Y0=y_substitution[0],
        Y1=y_substitution[1],
    ))
    return AffineChartRing.ideal([
        chart_equation,
        chart_equation.derivative(u),
        chart_equation.derivative(v),
    ])


chart_singularity_ideals = tuple(
    singular_ideal_on_chart(smooth_example, i, j)
    for i in (0, 1)
    for j in (0, 1)
)
smooth_example_is_invariant = tau_on_section(smooth_example) == smooth_example
smooth_example_is_smooth = all(
    ideal.is_one() for ideal in chart_singularity_ideals
)
smooth_example_fixed_values = tuple(
    evaluate_section_at_fixed_point(smooth_example, point)
    for point in fixed_coordinate_points
)
smooth_example_avoids_fixed_points = all(
    value != 0 for value in smooth_example_fixed_values
)

smooth_avoid_open_nonempty = (
    base_locus_is_empty
    and smooth_example_is_invariant
    and smooth_example_is_smooth
    and smooth_example_avoids_fixed_points
)

# Bertini gives a dense open smooth locus in P(V_+); intersecting it with
# the nonempty open avoidance locus remains a nonempty open sublocus.
parameter_space_dimension = P_V_plus_dimension
smooth_avoid_open_dimension = parameter_space_dimension
ade_dimension_lower_bound = smooth_avoid_open_dimension
ade_dimension_upper_bound = avoidance_locus_dimension

assert base_locus_is_empty
assert smooth_example_is_invariant
assert smooth_example_is_smooth
assert smooth_example_avoids_fixed_points
assert smooth_avoid_open_nonempty
assert ade_dimension_lower_bound == ade_dimension_upper_bound

ade_locus_dimension = ade_dimension_lower_bound

show(LatexExpr(r'I_{\mathrm{base}}^{\mathrm{sat}}='), saturated_base_ideal)
show(LatexExpr(r's_{\mathrm{example}}='), smooth_example)
show(
    LatexExpr(r'(s_{\mathrm{example}}(p_i))_{i=1}^4='),
    smooth_example_fixed_values,
)
print(
    'Jacobian ideals are unit ideals on all four affine charts:',
    smooth_example_is_smooth,
)
print(
    'The smooth fixed-point-avoiding open sublocus is nonempty:',
    smooth_avoid_open_nonempty,
)
show(
    LatexExpr(r'\dim\mathcal U_{\mathrm{smooth,avoid}}='),
    smooth_avoid_open_dimension,
)
show(
    LatexExpr(r'\dim\mathcal A\geq'),
    ade_dimension_lower_bound,
)
show(
    LatexExpr(r'\dim\mathcal A\leq'),
    ade_dimension_upper_bound,
)
show(
    LatexExpr(r'\dim\mathcal A='),
    ade_locus_dimension,
)
```

## What remains to compute the actual ADE locus

The preceding dimension argument uses only the open smooth sublocus. To determine the full at-worst-ADE locus itself, one would still have to perform the following calculations.

1. Introduce homogeneous parameters $a_{ij}$ for a universal invariant section

$$
F=\sum_{i+j\ \mathrm{even}} a_{ij}X_0^{4-i}X_1^iY_0^{4-j}Y_1^j
$$

over $Y\times\mathbf P(V_+)$.

2. On each standard affine chart of $Y$, compute the relative singular scheme of the universal curve from

$$
F=\frac{\partial F}{\partial u}=\frac{\partial F}{\partial v}=0.
$$

Eliminating the chart coordinates would produce equations for the discriminant in $\mathbf P(V_+)$, whose complement is the smooth locus.

3. Over the discriminant, compute the local Jacobian or Tjurina algebra at every singular point. One must first test that the singularity is isolated; positive-dimensional singular loci and nonreduced components are automatically outside the desired locus.

4. For each isolated singularity, compare the completed local equation with the normal forms

$$
A_n,\ D_n,\ E_6,\ E_7,\ E_8.
$$

A computational implementation would require finite-determinacy bounds and local coordinate changes, or an equivalent criterion using the local algebra together with additional tests distinguishing simple singularities from other isolated plane-curve singularities.

5. Impose the condition simultaneously at every singular point of the fiber, then project the resulting incidence strata to $\mathbf P(V_+)$. The union of the projected ADE strata, intersected with the fixed-point-avoidance open set, is the actual locus $\mathcal A$.

These elimination and local-classification steps have not been carried out here. The notebook presently proves only that $\mathcal A$ contains a nonempty open smooth sublocus of dimension $12$, and hence has dimension $12$ inside the $12$-dimensional avoidance locus.

## Universal local equations and the $A_1$ criterion

On each standard affine chart, write the universal invariant member as

$$
F(u,v)=0.
$$

For a plane-curve germ on a smooth surface over a field of characteristic different from $2$, the germ is of type $A_1$ when

$$
F=F_u=F_v=0,
\qquad
\det\operatorname{Hess}(F)\neq0.
$$

We compute the action of $(u,v)\mapsto(-u,-v)$ on $F$, its gradient, and its Hessian determinant on all four charts. We also compute the parameter ideal for a singularity at each chart origin and saturate it by the corresponding fixed-point-avoidance coefficient.

```python
UniversalChartRing = PolynomialRing(
    CoefficientRing,
    names=('u', 'v'),
)
u, v = UniversalChartRing.gens()


def universal_invariant_chart_equation(x_chart, y_chart):
    x_substitution = (
        (CoefficientRing(1), u)
        if x_chart == 0
        else (u, CoefficientRing(1))
    )
    y_substitution = (
        (CoefficientRing(1), v)
        if y_chart == 0
        else (v, CoefficientRing(1))
    )
    return UniversalChartRing(sum(
        coefficient_by_pair[(i, j)]
        * x_substitution[0]**(4-i)
        * x_substitution[1]**i
        * y_substitution[0]**(4-j)
        * y_substitution[1]**j
        for i, j in invariant_pairs
    ))


universal_A1_chart_data = []
for x_chart in (0, 1):
    for y_chart in (0, 1):
        F = universal_invariant_chart_equation(x_chart, y_chart)
        Fu = F.derivative(u)
        Fv = F.derivative(v)
        Hessian = matrix(UniversalChartRing, [
            [Fu.derivative(u), Fu.derivative(v)],
            [Fv.derivative(u), Fv.derivative(v)],
        ])
        Hessian_determinant = Hessian.det()

        section_is_invariant = F(u=-u, v=-v) == F
        gradient_is_anti_invariant = (
            Fu(u=-u, v=-v) == -Fu
            and Fv(u=-u, v=-v) == -Fv
        )
        Hessian_determinant_is_invariant = (
            Hessian_determinant(u=-u, v=-v)
            == Hessian_determinant
        )

        fixed_value = CoefficientRing(F(u=0, v=0))
        fixed_gradient = (
            CoefficientRing(Fu(u=0, v=0)),
            CoefficientRing(Fv(u=0, v=0)),
        )
        fixed_singularity_ideal = CoefficientRing.ideal([
            fixed_value,
            fixed_gradient[0],
            fixed_gradient[1],
        ])
        fixed_singularity_inside_avoidance = (
            fixed_singularity_ideal.saturation(
                CoefficientRing.ideal(fixed_value)
            )[0]
        )

        universal_A1_chart_data.append({
            'chart': (x_chart, y_chart),
            'F': F,
            'Fu': Fu,
            'Fv': Fv,
            'Hessian_determinant': Hessian_determinant,
            'section_is_invariant': section_is_invariant,
            'gradient_is_anti_invariant': gradient_is_anti_invariant,
            'Hessian_determinant_is_invariant': (
                Hessian_determinant_is_invariant
            ),
            'fixed_value': fixed_value,
            'fixed_singularity_inside_avoidance': (
                fixed_singularity_inside_avoidance
            ),
        })

universal_singularity_equations_are_tau_stable = all(
    data['section_is_invariant']
    and data['gradient_is_anti_invariant']
    for data in universal_A1_chart_data
)
universal_A1_hessian_condition_is_tau_stable = all(
    data['Hessian_determinant_is_invariant']
    for data in universal_A1_chart_data
)
fixed_singularity_inside_avoidance_is_empty = all(
    data['fixed_singularity_inside_avoidance'].is_one()
    for data in universal_A1_chart_data
)

assert universal_singularity_equations_are_tau_stable
assert universal_A1_hessian_condition_is_tau_stable
assert fixed_singularity_inside_avoidance_is_empty

show(
    LatexExpr(r'F_{00}(u,v)='),
    universal_A1_chart_data[0]['F'],
)
print(
    'Singularity equations are tau-stable on all four charts:',
    universal_singularity_equations_are_tau_stable,
)
print(
    'The nondegenerate-Hessian A1 condition is tau-stable:',
    universal_A1_hessian_condition_is_tau_stable,
)
print('Fixed-point chart data:')
for data in universal_A1_chart_data:
    print(
        data['chart'],
        'fixed value =',
        data['fixed_value'],
        'singular inside avoidance ideal =',
        data['fixed_singularity_inside_avoidance'],
    )
```

## The exact-one-$A_1$ locus

We use the following orbit argument. An automorphism preserving a divisor preserves its singular subscheme and its local analytic singularity type. Hence an $A_1$ point not fixed by $\tau$ occurs together with the distinct $A_1$ point in its $\tau$-orbit. A divisor with exactly one $A_1$ point would therefore have that point in $\operatorname{Fix}(\tau)$.

The preceding chart computation tests the remaining possibility: singularity at a fixed point after imposing the fixed-point-avoidance condition. We combine the orbit argument with the computed saturated fixed-point incidence ideals and record the resulting parameter-space ideal.

```python
orbit_pairing_for_A1_verified = (
    universal_singularity_equations_are_tau_stable
    and universal_A1_hessian_condition_is_tau_stable
)
exactly_one_A1_locus_is_empty = (
    orbit_pairing_for_A1_verified
    and fixed_singularity_inside_avoidance_is_empty
)
exactly_one_A1_locus_ideal = (
    CoefficientRing.ideal(1)
    if exactly_one_A1_locus_is_empty
    else CoefficientRing.ideal(0)
)

assert orbit_pairing_for_A1_verified
assert exactly_one_A1_locus_is_empty
assert exactly_one_A1_locus_ideal.is_one()

print(
    'A1 singularities away from the fixed locus occur in tau-orbits:',
    orbit_pairing_for_A1_verified,
)
print(
    'A fixed A1 point is compatible with the avoidance condition:',
    not fixed_singularity_inside_avoidance_is_empty,
)
show(
    LatexExpr(r'I_{\mathrm{exactly\ one}\ A_1}='),
    exactly_one_A1_locus_ideal,
)
print(
    'The invariant fixed-point-avoiding family with exactly one A1 point is empty:',
    exactly_one_A1_locus_is_empty,
)
```

## Dropping avoidance: one $A_1$ point and the fixed locus

We now remove the condition $B\cap\operatorname{Fix}(\tau)=\varnothing$ and impose instead that $B$ have exactly one singular point, of type $A_1$.

We use two inputs. First, an automorphism preserving a divisor preserves its singular subscheme and the completed local singularity type; hence the singular points occur in $\tau$-orbits. Second, an $A_1$ plane-curve germ is detected by vanishing of the equation and gradient together with a nonzero Hessian determinant.

The next cell computes the local parity and Hessian behavior of the universal sections in both eigenspaces at all four fixed points.

```python
corner_pair_by_chart = {
    (0, 0): (0, 0),
    (0, 1): (0, 4),
    (1, 0): (4, 0),
    (1, 1): (4, 4),
}
fixed_point_for_corner_pair = {
    (0, 0): ((QQ(1), QQ(0)), (QQ(1), QQ(0))),
    (0, 4): ((QQ(1), QQ(0)), (QQ(0), QQ(1))),
    (4, 0): ((QQ(0), QQ(1)), (QQ(1), QQ(0))),
    (4, 4): ((QQ(0), QQ(1)), (QQ(0), QQ(1))),
}

# Local data for the invariant eigenspace.
invariant_fixed_local_data = {}
for data in universal_A1_chart_data:
    chart = data['chart']
    pair = corner_pair_by_chart[chart]
    F = data['F']
    Fu = data['Fu']
    Fv = data['Fv']
    fixed_value = CoefficientRing(F(u=0, v=0))
    fixed_gradient = (
        CoefficientRing(Fu(u=0, v=0)),
        CoefficientRing(Fv(u=0, v=0)),
    )
    fixed_hessian_determinant = CoefficientRing(
        data['Hessian_determinant'](u=0, v=0)
    )
    invariant_fixed_local_data[pair] = {
        'chart': chart,
        'value': fixed_value,
        'gradient': fixed_gradient,
        'hessian_determinant': fixed_hessian_determinant,
    }

expected_hessian_by_corner = {
    (0, 0): 4 * coefficient_by_pair[(2, 0)] * coefficient_by_pair[(0, 2)]
            - coefficient_by_pair[(1, 1)]**2,
    (0, 4): 4 * coefficient_by_pair[(2, 4)] * coefficient_by_pair[(0, 2)]
            - coefficient_by_pair[(1, 3)]**2,
    (4, 0): 4 * coefficient_by_pair[(2, 0)] * coefficient_by_pair[(4, 2)]
            - coefficient_by_pair[(3, 1)]**2,
    (4, 4): 4 * coefficient_by_pair[(2, 4)] * coefficient_by_pair[(4, 2)]
            - coefficient_by_pair[(3, 3)]**2,
}

assert all(
    local_data['gradient'] == (0, 0)
    for local_data in invariant_fixed_local_data.values()
)
assert all(
    invariant_fixed_local_data[pair]['value']
    == coefficient_by_pair[pair]
    for pair in corner_pair_set
)
assert all(
    invariant_fixed_local_data[pair]['hessian_determinant']
    == expected_hessian_by_corner[pair]
    for pair in corner_pair_set
)

# Universal anti-invariant local equations.
anti_coefficient_names = [
    f'b{i}{j}' for i, j in anti_invariant_pairs
]
AntiCoefficientRing = PolynomialRing(QQ, names=anti_coefficient_names)
anti_coefficients = AntiCoefficientRing.gens()
anti_coefficient_by_pair = dict(
    zip(anti_invariant_pairs, anti_coefficients)
)
AntiChartRing = PolynomialRing(AntiCoefficientRing, names=('r', 's'))
r, s = AntiChartRing.gens()


def universal_anti_invariant_chart_equation(x_chart, y_chart):
    x_substitution = (
        (AntiCoefficientRing(1), r)
        if x_chart == 0
        else (r, AntiCoefficientRing(1))
    )
    y_substitution = (
        (AntiCoefficientRing(1), s)
        if y_chart == 0
        else (s, AntiCoefficientRing(1))
    )
    return AntiChartRing(sum(
        anti_coefficient_by_pair[(i, j)]
        * x_substitution[0]**(4-i)
        * x_substitution[1]**i
        * y_substitution[0]**(4-j)
        * y_substitution[1]**j
        for i, j in anti_invariant_pairs
    ))


anti_fixed_local_data = {}
for chart, pair in corner_pair_by_chart.items():
    F_minus = universal_anti_invariant_chart_equation(*chart)
    Fr = F_minus.derivative(r)
    Fs = F_minus.derivative(s)
    Hessian_minus = matrix(AntiChartRing, [
        [Fr.derivative(r), Fr.derivative(s)],
        [Fs.derivative(r), Fs.derivative(s)],
    ])
    hessian_determinant_at_fixed = AntiCoefficientRing(
        Hessian_minus.det()(r=0, s=0)
    )
    anti_fixed_local_data[pair] = {
        'chart': chart,
        'equation': F_minus,
        'value': AntiCoefficientRing(F_minus(r=0, s=0)),
        'hessian_determinant': hessian_determinant_at_fixed,
        'is_odd': F_minus(r=-r, s=-s) == -F_minus,
    }

assert all(
    data['is_odd']
    for data in anti_fixed_local_data.values()
)
assert all(
    data['value'] == 0
    for data in anti_fixed_local_data.values()
)
assert all(
    data['hessian_determinant'] == 0
    for data in anti_fixed_local_data.values()
)

print('Invariant fixed-point local data:')
for pair in sorted(corner_pair_set):
    data = invariant_fixed_local_data[pair]
    print(
        pair,
        'value =', data['value'],
        'gradient =', data['gradient'],
        'Hessian determinant =', data['hessian_determinant'],
    )
print('Anti-invariant fixed-point Hessian determinants:')
for pair in sorted(corner_pair_set):
    print(pair, anti_fixed_local_data[pair]['hessian_determinant'])
```

## Candidate coefficient strata

For each fixed point indexed by a corner pair $\alpha$, we form the candidate locally closed coefficient conditions

$$
a_\alpha=0,
\qquad
\prod_{\beta\neq\alpha}a_\beta\neq0,
\qquad
\Delta_\alpha\neq0,
$$

where $\Delta_\alpha$ is the Hessian determinant computed in the corresponding affine chart. The next cell combines these local conditions with the orbit calculation and records the additional global condition needed for “exactly one singularity”: the singular scheme away from the chosen fixed point must be empty.

```python
unique_A1_forces_fixed_point = orbit_pairing_for_A1_verified
anti_invariant_fixed_A1_possible = any(
    data['hessian_determinant'] != 0
    for data in anti_fixed_local_data.values()
)
invariant_fixed_incidence_forces_singularity = all(
    data['gradient'] == (0, 0)
    for data in invariant_fixed_local_data.values()
)

assert unique_A1_forces_fixed_point
assert not anti_invariant_fixed_A1_possible
assert invariant_fixed_incidence_forces_singularity

exactly_one_A1_forces_invariant_eigenline = (
    unique_A1_forces_fixed_point
    and not anti_invariant_fixed_A1_possible
)
exactly_one_A1_forces_exactly_one_fixed_point_on_curve = (
    exactly_one_A1_forces_invariant_eigenline
    and invariant_fixed_incidence_forces_singularity
)

candidate_one_A1_strata = {}
for pair in sorted(corner_pair_set):
    other_pairs = tuple(
        other_pair
        for other_pair in sorted(corner_pair_set)
        if other_pair != pair
    )
    candidate_one_A1_strata[pair] = {
        'vanishing_coefficient': coefficient_by_pair[pair],
        'other_corner_product': prod(
            coefficient_by_pair[other_pair]
            for other_pair in other_pairs
        ),
        'hessian_determinant': (
            invariant_fixed_local_data[pair]['hessian_determinant']
        ),
        'global_condition': (
            'the singular scheme away from the chosen fixed point is empty'
        ),
    }

assert all(
    data['hessian_determinant'] != 0
    for data in candidate_one_A1_strata.values()
)

print(
    'Uniqueness of an A1 point forces the point to be tau-fixed:',
    unique_A1_forces_fixed_point,
)
print(
    'An anti-invariant section can have an A1 point at a fixed point:',
    anti_invariant_fixed_A1_possible,
)
print(
    'For invariant sections, incidence with a fixed point forces singularity:',
    invariant_fixed_incidence_forces_singularity,
)
print(
    'Exactly one A1 point forces exactly one fixed point on the curve:',
    exactly_one_A1_forces_exactly_one_fixed_point_on_curve,
)
print('Candidate coefficient strata:')
for pair, data in candidate_one_A1_strata.items():
    print(
        pair,
        ':',
        data['vanishing_coefficient'], '= 0,',
        data['other_corner_product'], '!= 0,',
        data['hessian_determinant'], '!= 0;',
        data['global_condition'],
    )
```

## A verified member with exactly one $A_1$ point

We test nonemptiness of one candidate stratum by choosing a section with the coefficient of $X_0^4Y_0^4$ equal to zero, the other three corner coefficients nonzero, and nondegenerate quadratic part at $([1:0],[1:0])$.

On each standard affine chart we compute the singular ideal generated by the equation and its two first derivatives. Exactness requires the singular ideal to be the maximal ideal of the chosen chart origin and to be the unit ideal on the other three charts.

```python
one_A1_example = (
    X0**2 * X1**2 * Y0**4
    + X0**4 * Y0**2 * Y1**2
    + X0**4 * Y1**4
    + X1**4 * Y0**4
    + X1**4 * Y1**4
)

one_A1_example_fixed_values = {
    pair: evaluate_section_at_fixed_point(
        one_A1_example,
        fixed_point_for_corner_pair[pair],
    )
    for pair in corner_pair_set
}
zero_fixed_pair_set = frozenset(
    pair
    for pair, value in one_A1_example_fixed_values.items()
    if value == 0
)

one_A1_example_chart_data = {}
for chart, pair in corner_pair_by_chart.items():
    AffineChartRing = PolynomialRing(QQ, names=('w', 'z'))
    w, z = AffineChartRing.gens()
    x_substitution = (
        (QQ(1), w)
        if chart[0] == 0
        else (w, QQ(1))
    )
    y_substitution = (
        (QQ(1), z)
        if chart[1] == 0
        else (z, QQ(1))
    )
    chart_equation = AffineChartRing(one_A1_example(
        X0=x_substitution[0],
        X1=x_substitution[1],
        Y0=y_substitution[0],
        Y1=y_substitution[1],
    ))
    chart_gradient = (
        chart_equation.derivative(w),
        chart_equation.derivative(z),
    )
    chart_hessian = matrix(AffineChartRing, [
        [
            chart_gradient[0].derivative(w),
            chart_gradient[0].derivative(z),
        ],
        [
            chart_gradient[1].derivative(w),
            chart_gradient[1].derivative(z),
        ],
    ])
    chart_singular_ideal = AffineChartRing.ideal([
        chart_equation,
        chart_gradient[0],
        chart_gradient[1],
    ])
    one_A1_example_chart_data[pair] = {
        'chart': chart,
        'equation': chart_equation,
        'singular_ideal': chart_singular_ideal,
        'hessian_determinant_at_origin': QQ(
            chart_hessian.det()(w=0, z=0)
        ),
        'origin_maximal_ideal': AffineChartRing.ideal([w, z]),
    }

chosen_pair = (0, 0)
chosen_chart_data = one_A1_example_chart_data[chosen_pair]
other_chart_data = [
    data
    for pair, data in one_A1_example_chart_data.items()
    if pair != chosen_pair
]

assert tau_on_section(one_A1_example) == one_A1_example
assert zero_fixed_pair_set == frozenset({chosen_pair})
assert (
    chosen_chart_data['singular_ideal']
    == chosen_chart_data['origin_maximal_ideal']
)
assert chosen_chart_data['hessian_determinant_at_origin'] != 0
assert all(
    data['singular_ideal'].is_one()
    for data in other_chart_data
)

one_A1_example_has_exactly_one_A1 = (
    zero_fixed_pair_set == frozenset({chosen_pair})
    and chosen_chart_data['singular_ideal']
       == chosen_chart_data['origin_maximal_ideal']
    and chosen_chart_data['hessian_determinant_at_origin'] != 0
    and all(
        data['singular_ideal'].is_one()
        for data in other_chart_data
    )
)
assert one_A1_example_has_exactly_one_A1

show(LatexExpr(r's_{1A_1}='), one_A1_example)
print('Values at the four fixed points, indexed by corner pairs:')
for pair in sorted(corner_pair_set):
    print(pair, one_A1_example_fixed_values[pair])
print(
    'Singular ideal at the chosen fixed point:',
    chosen_chart_data['singular_ideal'],
)
show(
    LatexExpr(r'\det\operatorname{Hess}_{p_{00}}(s_{1A_1})='),
    chosen_chart_data['hessian_determinant_at_origin'],
)
print('Singular ideals on the other affine charts:')
for pair in sorted(corner_pair_set - {chosen_pair}):
    print(pair, one_A1_example_chart_data[pair]['singular_ideal'])
print(
    'The section has exactly one A1 singularity:',
    one_A1_example_has_exactly_one_A1,
)
```

## Exact one-node locus and its dimension

Let $P=\mathbf P(V_+)\cong\mathbf P^{12}$. For each corner index $\alpha\in\{00,04,40,44\}$, let $p_\alpha$ be the corresponding fixed point and let $H_\alpha\subset P$ be the hyperplane $a_\alpha=0$.

We compute the local open condition on $H_\alpha$ requiring:

$$
\prod_{\beta\neq\alpha}a_\beta\neq0
$$

and nonvanishing of the Hessian determinant at $p_\alpha$.

To encode the global condition, let $\Sigma\subset Y\times P$ be the universal singular scheme. Over $H_\alpha$, remove the persistent fixed-point section $\{p_\alpha\}\times H_\alpha$ by saturation, obtaining the residual singular incidence scheme $R_\alpha$. Let

$$
D_\alpha^{\mathrm{extra}}=\pi(R_\alpha)\subset H_\alpha,
$$

where $\pi:Y\times P\to P$ is the projective projection. The exact stratum is defined by

$$
\mathcal N_\alpha=
H_\alpha\cap
D_+\!\left(
\Delta_\alpha\prod_{\beta\neq\alpha}a_\beta
\right)
\setminus D_\alpha^{\mathrm{extra}}.
$$

The next cell computes the hyperplane dimensions, the four local factors, and explicit members used to test nonemptiness. The expanded generators of $D_\alpha^{\mathrm{extra}}$ are not computed here; they are defined exactly by the residual incidence and proper projection above.

```python
from sage.all import *

S_node = PolynomialRing(QQ, names=('X0', 'X1', 'Y0', 'Y1'))
X0, X1, Y0, Y1 = S_node.gens()

all_pairs_node = tuple(
    (i, j) for i in range(5) for j in range(5)
)
invariant_pairs_node = tuple(
    (i, j) for i, j in all_pairs_node if (i + j) % 2 == 0
)
corner_pairs_node = frozenset({
    (0, 0), (0, 4), (4, 0), (4, 4),
})

coefficient_names_node = tuple(
    f'a{i}{j}' for i, j in invariant_pairs_node
)
ParameterSpace_node = ProjectiveSpace(
    QQ,
    len(invariant_pairs_node) - 1,
    names=coefficient_names_node,
)
parameter_coordinates_node = ParameterSpace_node.gens()
parameter_coordinate_by_pair_node = dict(
    zip(invariant_pairs_node, parameter_coordinates_node)
)

hessian_by_corner_node = {
    (0, 0): (
        4 * parameter_coordinate_by_pair_node[(0, 2)]
          * parameter_coordinate_by_pair_node[(2, 0)]
        - parameter_coordinate_by_pair_node[(1, 1)]**2
    ),
    (0, 4): (
        4 * parameter_coordinate_by_pair_node[(0, 2)]
          * parameter_coordinate_by_pair_node[(2, 4)]
        - parameter_coordinate_by_pair_node[(1, 3)]**2
    ),
    (4, 0): (
        4 * parameter_coordinate_by_pair_node[(2, 0)]
          * parameter_coordinate_by_pair_node[(4, 2)]
        - parameter_coordinate_by_pair_node[(3, 1)]**2
    ),
    (4, 4): (
        4 * parameter_coordinate_by_pair_node[(2, 4)]
          * parameter_coordinate_by_pair_node[(4, 2)]
        - parameter_coordinate_by_pair_node[(3, 3)]**2
    ),
}

node_hyperplanes = {
    pair: ParameterSpace_node.subscheme([
        parameter_coordinate_by_pair_node[pair]
    ])
    for pair in corner_pairs_node
}
node_hyperplane_dimensions = {
    pair: hyperplane.dimension()
    for pair, hyperplane in node_hyperplanes.items()
}
node_local_open_factors = {
    pair: (
        hessian_by_corner_node[pair]
        * prod(
            parameter_coordinate_by_pair_node[other_pair]
            for other_pair in corner_pairs_node
            if other_pair != pair
        )
    )
    for pair in corner_pairs_node
}

assert ParameterSpace_node.dimension() == 12
assert set(node_hyperplane_dimensions.values()) == {11}
assert all(
    factor != 0 for factor in node_local_open_factors.values()
)

fixed_point_by_corner_node = {
    (0, 0): ((QQ(1), QQ(0)), (QQ(1), QQ(0))),
    (0, 4): ((QQ(1), QQ(0)), (QQ(0), QQ(1))),
    (4, 0): ((QQ(0), QQ(1)), (QQ(1), QQ(0))),
    (4, 4): ((QQ(0), QQ(1)), (QQ(0), QQ(1))),
}
chart_by_corner_node = {
    (0, 0): (0, 0),
    (0, 4): (0, 1),
    (4, 0): (1, 0),
    (4, 4): (1, 1),
}

base_one_node_section = (
    X0**2 * X1**2 * Y0**4
    + X0**4 * Y0**2 * Y1**2
    + X0**4 * Y1**4
    + X1**4 * Y0**4
    + X1**4 * Y1**4
)


def swap_projective_coordinates(section, swap_x=False, swap_y=False):
    return S_node(section(
        X0=X1 if swap_x else X0,
        X1=X0 if swap_x else X1,
        Y0=Y1 if swap_y else Y0,
        Y1=Y0 if swap_y else Y1,
    ))


example_by_corner_node = {
    (0, 0): swap_projective_coordinates(base_one_node_section),
    (0, 4): swap_projective_coordinates(
        base_one_node_section,
        swap_y=True,
    ),
    (4, 0): swap_projective_coordinates(
        base_one_node_section,
        swap_x=True,
    ),
    (4, 4): swap_projective_coordinates(
        base_one_node_section,
        swap_x=True,
        swap_y=True,
    ),
}


def evaluate_node_section(section, point):
    (x0, x1), (y0, y1) = point
    return QQ(section(X0=x0, X1=x1, Y0=y0, Y1=y1))


def verify_exactly_one_A1(section, chosen_pair):
    tau_section = S_node(section(
        X0=X0,
        X1=-X1,
        Y0=Y0,
        Y1=-Y1,
    ))
    fixed_values = {
        pair: evaluate_node_section(section, point)
        for pair, point in fixed_point_by_corner_node.items()
    }
    zero_fixed_pairs = frozenset(
        pair for pair, value in fixed_values.items() if value == 0
    )

    chart_results = {}
    for pair, chart in chart_by_corner_node.items():
        A = PolynomialRing(QQ, names=('u_node', 'v_node'))
        u_node, v_node = A.gens()
        x_substitution = (
            (QQ(1), u_node)
            if chart[0] == 0
            else (u_node, QQ(1))
        )
        y_substitution = (
            (QQ(1), v_node)
            if chart[1] == 0
            else (v_node, QQ(1))
        )
        f = A(section(
            X0=x_substitution[0],
            X1=x_substitution[1],
            Y0=y_substitution[0],
            Y1=y_substitution[1],
        ))
        fu = f.derivative(u_node)
        fv = f.derivative(v_node)
        singular_ideal = A.ideal([f, fu, fv])
        hessian = matrix(A, [
            [fu.derivative(u_node), fu.derivative(v_node)],
            [fv.derivative(u_node), fv.derivative(v_node)],
        ])
        chart_results[pair] = {
            'singular_ideal': singular_ideal,
            'origin_ideal': A.ideal([u_node, v_node]),
            'hessian_at_origin': QQ(
                hessian.det()(u_node=0, v_node=0)
            ),
        }

    chosen_data = chart_results[chosen_pair]
    other_data = [
        data for pair, data in chart_results.items()
        if pair != chosen_pair
    ]
    verified = (
        tau_section == section
        and zero_fixed_pairs == frozenset({chosen_pair})
        and chosen_data['singular_ideal']
            == chosen_data['origin_ideal']
        and chosen_data['hessian_at_origin'] != 0
        and all(data['singular_ideal'].is_one() for data in other_data)
    )
    return verified, fixed_values, chart_results


example_verification_node = {
    pair: verify_exactly_one_A1(section, pair)
    for pair, section in example_by_corner_node.items()
}
node_strata_nonempty = all(
    result[0] for result in example_verification_node.values()
)

assert node_strata_nonempty

node_stratum_dimension = 11
exactly_one_A1_locus_dimension = max(
    node_hyperplane_dimensions.values()
)
assert exactly_one_A1_locus_dimension == node_stratum_dimension

show(LatexExpr(r'\dim P='), ParameterSpace_node.dimension())
print('Hyperplane dimensions:')
for pair in sorted(corner_pairs_node):
    print(pair, node_hyperplane_dimensions[pair])
print('Local open factors:')
for pair in sorted(corner_pairs_node):
    print(pair, node_local_open_factors[pair])
print('Verified one-node examples:')
for pair in sorted(corner_pairs_node):
    verified, fixed_values, chart_results = example_verification_node[pair]
    print(pair, 'verified =', verified, 'fixed values =', fixed_values)
show(LatexExpr(r'\dim\mathcal N_\alpha='), node_stratum_dimension)
show(
    LatexExpr(r'\dim\mathcal N_{\mathrm{exactly\ one}\ A_1}='),
    exactly_one_A1_locus_dimension,
)
print(
    'Expanded residual-discriminant generators were not computed; '
    'the exact global exclusion is represented by the residual '
    'singular incidence projection.'
)
```

## Testing the proposed converse

We test whether meeting exactly one $\tau$-fixed point forces the unique singularity to be $A_1$. For a local equation $f(u,v)$ with $f_{uu}(0,0)\neq0$, the formal splitting lemma reduces the germ to

$$
U^2+g(v).
$$

If $g$ has order $m$, the resulting plane-curve singularity has type $A_{m-1}$. The following cell computes the fixed-point incidence, the global singular scheme on all four affine charts, and the splitting data for an explicit invariant member.

```python
from sage.all import *

S_counterexample = PolynomialRing(QQ, names=('X0', 'X1', 'Y0', 'Y1'))
X0, X1, Y0, Y1 = S_counterexample.gens()

B_A3 = (
    X0**2 * X1**2 * Y0**4
    + X0**4 * Y1**4
    + X1**4 * Y0**4
    + X1**4 * Y1**4
)

fixed_points_counterexample = {
    (0, 0): ((QQ(1), QQ(0)), (QQ(1), QQ(0))),
    (0, 4): ((QQ(1), QQ(0)), (QQ(0), QQ(1))),
    (4, 0): ((QQ(0), QQ(1)), (QQ(1), QQ(0))),
    (4, 4): ((QQ(0), QQ(1)), (QQ(0), QQ(1))),
}


def tau_counterexample(section):
    return section(X0=X0, X1=-X1, Y0=Y0, Y1=-Y1)


def evaluate_counterexample(section, point):
    (a0, a1), (b0, b1) = point
    return QQ(section(X0=a0, X1=a1, Y0=b0, Y1=b1))


fixed_values_counterexample = {
    pair: evaluate_counterexample(B_A3, point)
    for pair, point in fixed_points_counterexample.items()
}
fixed_points_on_curve_counterexample = frozenset(
    pair
    for pair, value in fixed_values_counterexample.items()
    if value == 0
)

chart_by_corner_counterexample = {
    (0, 0): (0, 0),
    (0, 4): (0, 1),
    (4, 0): (1, 0),
    (4, 4): (1, 1),
}
chart_data_counterexample = {}

for pair, chart in chart_by_corner_counterexample.items():
    A = PolynomialRing(QQ, names=('u', 'v'))
    u, v = A.gens()
    x_substitution = (
        (QQ(1), u)
        if chart[0] == 0
        else (u, QQ(1))
    )
    y_substitution = (
        (QQ(1), v)
        if chart[1] == 0
        else (v, QQ(1))
    )
    local_equation = A(B_A3(
        X0=x_substitution[0],
        X1=x_substitution[1],
        Y0=y_substitution[0],
        Y1=y_substitution[1],
    ))
    local_gradient = (
        local_equation.derivative(u),
        local_equation.derivative(v),
    )
    singular_ideal = A.ideal([
        local_equation,
        local_gradient[0],
        local_gradient[1],
    ])
    chart_data_counterexample[pair] = {
        'ring': A,
        'u': u,
        'v': v,
        'equation': local_equation,
        'gradient': local_gradient,
        'singular_ideal': singular_ideal,
    }

chosen_pair_counterexample = (0, 0)
chosen_data_counterexample = chart_data_counterexample[
    chosen_pair_counterexample
]
A_local = chosen_data_counterexample['ring']
u_local = chosen_data_counterexample['u']
v_local = chosen_data_counterexample['v']
f_local = chosen_data_counterexample['equation']
fu_local = chosen_data_counterexample['gradient'][0]

fu_over_u = A_local(fu_local // u_local)
fu_over_u_constant = QQ(
    fu_over_u.subs({u_local: 0, v_local: 0})
)
residual_one_variable = A_local(
    f_local.subs({u_local: 0})
)
residual_support = [
    exponent[1]
    for exponent, coefficient in residual_one_variable.dict().items()
    if coefficient != 0
]
residual_order = min(residual_support)
computed_A_index = residual_order - 1

tjurina_dimension = chosen_data_counterexample[
    'singular_ideal'
].vector_space_dimension()

other_chart_singular_ideals_are_units = all(
    data['singular_ideal'].is_one()
    for pair, data in chart_data_counterexample.items()
    if pair != chosen_pair_counterexample
)
chosen_singular_support_is_origin = (
    chosen_data_counterexample['singular_ideal'].radical()
    == A_local.ideal([u_local, v_local])
)

assert tau_counterexample(B_A3) == B_A3
assert fixed_points_on_curve_counterexample == frozenset({(0, 0)})
assert chosen_singular_support_is_origin
assert other_chart_singular_ideals_are_units
assert fu_over_u_constant != 0
assert residual_order == 4
assert computed_A_index == 3
assert tjurina_dimension == 3

proposed_converse_holds = computed_A_index == 1
assert not proposed_converse_holds

show(LatexExpr(r'B_{A_3}='), B_A3)
print('Values at the four fixed points:')
for pair in sorted(fixed_points_counterexample):
    print(pair, fixed_values_counterexample[pair])
print('Fixed points lying on the curve:', fixed_points_on_curve_counterexample)
show(LatexExpr(r'f(u,v)='), f_local)
show(LatexExpr(r'f_u/u='), fu_over_u)
show(LatexExpr(r'f(0,v)='), residual_one_variable)
show(LatexExpr(r'\operatorname{ord}_0 f(0,v)='), residual_order)
show(LatexExpr(r'\tau_{\mathrm{Tjurina}}='), tjurina_dimension)
show(LatexExpr(r'\text{computed ADE type}=A_'), computed_A_index)
print('Singular ideal at the chosen fixed point:', chosen_data_counterexample['singular_ideal'])
print('Other chart singular ideals are unit ideals:', other_chart_singular_ideals_are_units)
print('The proposed converse holds:', proposed_converse_holds)
```
