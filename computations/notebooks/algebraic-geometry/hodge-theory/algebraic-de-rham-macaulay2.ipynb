# Algebraic de Rham cohomology with Macaulay2 Dmodules

```python
%display latex
algebras = Algebras(QQ).Associative().Unital().Commutative()
```

$$
\begin{align*}
H^*(\AA^n) &= k[0] \\
H^*(\PP^n) &= k[0] \oplus \bigoplus_{\substack{ i=1 \\ \text{even} } }^n k[i]
\end{align*}
$$

```python
R = QQ.polynomial_ring(("x", "y"))
x, y = R.algebra_generators()
f = y^3 - x^2
complement = R.localization(f)
r = algebras.de_rham_cohomology_algebra()(complement)
```
