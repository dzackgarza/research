# Generators of $H^0(\mathbb P^1\times\mathbb P^2,\mathcal O(5,5))$

Let $[x_0:x_1]$ be homogeneous coordinates on $\mathbb P^1$, and let $[y_0:y_1:y_2]$ be homogeneous coordinates on $\mathbb P^2$. By the Künneth isomorphism,

$$
H^0(\mathbb P^1\times\mathbb P^2,\mathcal O(5,5))
\cong H^0(\mathbb P^1,\mathcal O(5))\otimes H^0(\mathbb P^2,\mathcal O(5)).
$$

A basis consists of the bihomogeneous monomials

$$
x_0^{5-i}x_1^i y_0^a y_1^b y_2^c,
\qquad 0\le i\le 5,\quad a,b,c\ge 0,\quad a+b+c=5.
$$

The following SageMath cells construct this basis and verify

$$
\dim H^0(\mathbb P^1\times\mathbb P^2,\mathcal O(5,5))
=(5+1)\binom{5+2}{2}=126.
$$

```python
P1 = ProjectiveSpaces(QQ)(1)
P2 = ProjectiveSpaces(QQ)(2)
X = Schemes(QQ).product((P1, P2))
sections = X.O(5, 5).homogeneous_polynomial_sections()

R = sections.homogeneous_coordinate_ring()
x0, x1, y0, y1, y2 = R.algebra_generators()
show(R)
```

```python
a = 5
b = 5

basis = [x0^(a-i) * x1^i * y0^j * y1^k * y2^(b-j-k)
         for i in range(a + 1)
         for j in range(b + 1)
         for k in range(b - j + 1)]

show(basis)
```

```python
def bidegree_of_monomial(m):
    e = m.exponents()[0]
    return (e[0] + e[1], e[2] + e[3] + e[4])

expected_dimension = (a + 1) * ZZ(b + 2).binomial(2)

assert expected_dimension == 126
assert len(basis) == expected_dimension
assert len(set(basis)) == expected_dimension
assert all(bidegree_of_monomial(m) == (5, 5) for m in basis)
assert sections.dimension() == expected_dimension

sections.dimension()
```
