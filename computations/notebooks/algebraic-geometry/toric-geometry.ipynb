```python
# Create a list of vectors in R^2 representing the generators of the toric variety
vectors = [(1, 0), (0, 1), (-1, -1)]

# The rays of the fan of P^2; its maximal cones are the pairs of adjacent rays
fan = RationalPolyhedralFans(N)(((vectors[0], vectors[1]), (vectors[1], vectors[2]), (vectors[2], vectors[0])))

# Create a toric variety from the fan
variety = fan.toric_variety(QQ)
```

```python
# Compute the self-intersection numbers of the toric variety
rays = tuple(variety.fan().cones(1))
intersection_numbers = [variety.divisor_intersection(variety.torus_invariant_prime_divisor(rays[i]), variety.torus_invariant_prime_divisor(rays[i])) for i in range(len(rays))]

# Print the self-intersection numbers
print(intersection_numbers)
```

## Toric geometry: polytopes and invariant divisors

```python
# Intersection matrix of T-invariant divisors in P2
points = [
    (0, 0), (1, 0), (0, 1)
]
P2 = LatticePolytopes(N)(points)
show(P2)
P2_fan = P2.normal_fan()
P2_var = P2_fan.toric_variety(QQ)
show( P2_var.plot() )

rays = tuple(P2_fan.cones(1))
Ma = QQ.matrix_space(3, 3).from_rows([ [ P2_var.divisor_intersection(P2_var.torus_invariant_prime_divisor(rays[i]), P2_var.torus_invariant_prime_divisor(rays[j]))
           for i in range(0,3) ] for j in range(0,3) ])
show(Ma)
```

```python
class piso:
    def __init__(self, vertices = [ [0,0] ]):
        self.vertices = vertices
        self.poly = ConvexPolytopes(N)(self.vertices)
    def __repr__(self):
        return self.poly
    def __eq__(self, other):
        return self.poly.is_combinatorially_isomorphic(other.poly)

s = piso([ [1,1], [1,2], [-1,1] ])
s.poly
```

```python
class Fooset(set):
    def __init__(self, s=(), foo=None):
        super(Fooset,self).__init__(s)
        if foo is None and hasattr(s, 'foo'):
            foo = s.foo
        self.foo = foo
```
