# Primitive orbit coefficients for indefinite lattices

For a positive-definite integral lattice \\(L,b\\), put \\(Q(x):=b(x,x)\\). The ordinary theta series

\\[
\Theta_L(q)
=
\sum_{x\in L}q^{Q(x)}
\\]

has finite representation numbers as coefficients. For an indefinite lattice, a fixed nonzero norm shell is generally infinite and the represented norms are unbounded in both signs. The invariant studied here therefore separates positive norm, zero norm, and negative norm.

Let \\(R\\) be a Dedekind domain. An \\(R\\)-lattice is a finite projective \\(R\\)-module \\(L\\) equipped with a symmetric \\(R\\)-bilinear form

\\[
b:L\times L\longrightarrow R.
\\]

Put

\\[
L^\vee:=\operatorname{Hom}_R(L,R).
\\]

The left and right adjoints are

\\[
b^\flat_{\mathrm L}:L\longrightarrow L^\vee,
\qquad
x\longmapsto b(x,-),
\\]

and

\\[
b^\flat_{\mathrm R}:L\longrightarrow L^\vee,
\qquad
y\longmapsto b(-,y).
\\]

The form is nondegenerate when

\\[
\ker b^\flat_{\mathrm L}
=
\ker b^\flat_{\mathrm R}
=
0.
\\]

For a nondegenerate \\(R\\)-lattice define

\\[
O(L)
:=
\left\{
g\in\operatorname{Aut}_R(L)
\;\middle|\;
b(gx,gy)=b(x,y)\quad\forall x,y\in L
\right\}.
\\]

Let \\(\operatorname{Proj}^{\mathrm{fg}}(R)\\) denote the full subcategory of finitely generated projective \\(R\\)-modules. For \\(M,N\in\operatorname{Proj}^{\mathrm{fg}}(R)\\), set

\\[
\operatorname{AdmMono}_R(M,N)
:=
\left\{
\iota\in\operatorname{Hom}_R(M,N)
\;\middle|\;
\ker\iota=0,\ 
\operatorname{coker}(\iota)\in\operatorname{Proj}^{\mathrm{fg}}(R)
\right\}.
\\]

If \\(M,L\\) are \\(R\\)-lattices, define

\\[
\operatorname{Emb}^{\mathrm{prim}}(M,L)
:=
\left\{
\iota\in\operatorname{AdmMono}_R(M,L)
\;\middle|\;
b_L(\iota x,\iota y)=b_M(x,y)\quad\forall x,y\in M
\right\}.
\\]

Thus primitiveness is a property of the embedding \\(\iota\\), namely admissibility in the exact category of finite projective \\(R\\)-modules.

From now on \\(R=\mathbf Z\\). Write \\(Q(x):=b(x,x)\\). Let \\(L\\) have rank \\(r\\), and let the signature of \\(L_{\mathbf R}\\) be

\\[
(r_+,r_-),
\qquad
r_+r_->0.
\\]

Let \\(\Gamma\le O(L)\\) have finite index. For \\(n>0\\), let \\(\langle n\rangle\\) and \\(\langle-n\rangle\\) be rank-one lattices generated by \\(u_+\\) and \\(u_-\\) with

\\[
Q(u_+)=n,
\qquad
Q(u_-)=-n.
\\]

Define

\\[
C^+_{L,\Gamma}(n)
:=
\#
\left(
\Gamma\backslash
\operatorname{Emb}^{\mathrm{prim}}(\langle n\rangle,L)
\right)
\\]

and

\\[
C^-_{L,\Gamma}(n)
:=
\#
\left(
\Gamma\backslash
\operatorname{Emb}^{\mathrm{prim}}(\langle-n\rangle,L)
\right).
\\]

These cardinalities are finite. Let \\(V=L_{\mathbf Q}\\), let \\(m\ne0\\), and let \\(x,y\in V\\) satisfy \\(Q(x)=Q(y)=m\\). For nonisotropic \\(v\\), define

\\[
s_v(z)
:=
z-\frac{2b(z,v)}{Q(v)}v.
\\]

If \\(Q(x-y)\ne0\\), then \\(s_{x-y}(x)=y\\). If \\(Q(x-y)=0\\), then \\(Q(x+y)=4m\ne0\\) and \\(-s_{x+y}(x)=y\\). Hence \\(V(Q-m)\\) is one closed \\(O(V)\\)-orbit. A. Borel, *Arithmetic properties of linear algebraic groups*, Proc. ICM Stockholm 1962, §3(b), citing Borel--Harish-Chandra, *Arithmetic subgroups of algebraic groups*, Ann. of Math. (2) **75** (1962), Theorem 6.9, gives finiteness of arithmetic-group orbits on the integral points of such a closed rational orbit. Finite index gives the same conclusion for \\(\Gamma\\).

Set

\\[
\Theta^+_{L,\Gamma}(q)
:=
\sum_{n>0}
C^+_{L,\Gamma}(n)q^n
\in\mathbf Z[[q]]
\\]

and

\\[
\Theta^-_{L,\Gamma}(q)
:=
\sum_{n>0}
C^-_{L,\Gamma}(n)q^n
\in\mathbf Z[[q]].
\\]

Define the primitive isotropic monomorphisms by

\\[
\operatorname{Iso}^{\mathrm{prim}}(L)
:=
\left\{
\iota\in\operatorname{AdmMono}_{\mathbf Z}(\mathbf Z,L)
\;\middle|\;
Q(\iota(1))=0
\right\},
\\]

and set

\\[
\mathscr C^0_{L,\Gamma}
:=
\Gamma\backslash
\operatorname{Iso}^{\mathrm{prim}}(L).
\\]

Finally,

\\[
\mathcal T_{L,\Gamma}
:=
\left(
\Theta^+_{L,\Gamma},
\mathscr C^0_{L,\Gamma},
\Theta^-_{L,\Gamma}
\right).
\\]

The mass-defect method below concerns \\(C^+_{L,\Gamma}\\) and \\(C^-_{L,\Gamma}\\).

## 1. Stabilizers and the representation mass

For \\(m\ne0\\), set

\\[
X_m(L)
:=
\left\{
x\in L
\;\middle|\;
Q(x)=m
\right\}.
\\]

Let

\\[
V:=L_{\mathbf Q},
\qquad
G:=O(V).
\\]

For \\(x\in X_m(L)\\), define

\\[
W_x
:=
(\mathbf Qx)^{\perp V}
=
\left\{
v\in V
\;\middle|\;
b(v,x)=0
\right\}.
\\]

Since \\(m=Q(x)\ne0\\),

\\[
V
=
\mathbf Qx\perp W_x.
\\]

Let

\\[
G_x
:=
\operatorname{Stab}_{G}(x).
\\]

**Lemma.** Restriction to \\(W_x\\) induces an isomorphism

\\[
\operatorname{res}_x:
G_x
\xrightarrow{\sim}
O(W_x).
\\]

**Proof.** Every \\(g\in G_x\\) fixes \\(\mathbf Qx\\) pointwise and preserves \\(W_x\\), so restriction defines \\(\operatorname{res}_x\\). Conversely, if \\(h\in O(W_x)\\), then

\\[
\operatorname{id}_{\mathbf Qx}\perp h
\\]

is an isometry of \\(V=\mathbf Qx\perp W_x\\) fixing \\(x\\). The two constructions are inverse.

For the integral arithmetic group, set

\\[
O(L)_x
:=
\operatorname{Stab}_{O(L)}(x),
\qquad
\Gamma_x
:=
\operatorname{Stab}_{\Gamma}(x).
\\]

Thus \\(O(L)_x\\) and \\(\Gamma_x\\) are arithmetic subgroups of \\(G_x\simeq O(W_x)\\).

Fix Haar measures on \\(G(\mathbf R)\\) and \\(G_x(\mathbf R)\\) satisfying the normalization used by Eskin--Rudnick--Sarnak in equation (1.3) of *A proof of Siegel's weight formula*. Define

\\[
w_\Gamma(x)
:=
\frac{
\operatorname{vol}\!\left(
\Gamma_x\backslash G_x(\mathbf R)
\right)
}{
\operatorname{vol}\!\left(
\Gamma\backslash G(\mathbf R)
\right)
}.
\\]

For a complete set of representatives \\(x_1,\ldots,x_h\\) for \\(\Gamma\backslash X_m(L)\\), define

\\[
\mathcal M_{L,\Gamma}(m)
:=
\sum_{i=1}^{h} w_\Gamma(x_i).
\\]

## 2. Eskin--Rudnick--Sarnak's weight formula

The mass identity used below is the rank-one specialization of equation (1.5) in:

A. Eskin, Z. Rudnick, P. Sarnak, *A proof of Siegel's weight formula*, International Mathematics Research Notices **1991** (1991), no. 5, 65--69, DOI 10.1155/S1073792891000090.

They consider the integral matrix equation

\\[
{}^tXAX=B
\\]

for nondegenerate symmetric half-integral matrices \\(A\\) and \\(B\\), of sizes \\(r\\) and \\(s\\). Let \\(G=O(A)\\), let \\(X_1,\ldots,X_h\\) represent the \\(G(\mathbf Z)\\)-orbits of integral solutions, and let \\(G_{X_i}\\) be the stabilizer of \\(X_i\\). Under their condition (1.4),

\\[
s<\min(r_+,r_-),
\qquad
2s+2<r,
\\]

their equation (1.5) is

\\[
\frac{
\displaystyle
\sum_{i=1}^{h}
\operatorname{vol}\!\left(
G_{X_i}(\mathbf Z)\backslash G_{X_i}(\mathbf R)
\right)
}{
\operatorname{vol}\!\left(
G(\mathbf Z)\backslash G(\mathbf R)
\right)
}
=
\prod_p \alpha_p(A,B).
\\]

The Haar measures are those fixed in their equation (1.3). Their local density is equation (1.6),

\\[
\alpha_p(A,B)
:=
\lim_{a\to\infty}
p^{-ad}
\#
\left\{
X\bmod p^a
\;\middle|\;
{}^tXAX\equiv B\pmod{p^a}
\right\},
\qquad
d=rs-\frac{s(s+1)}2.
\\]

For a rank-one target \\(s=1\\), \\(B=(m)\\), one has \\(d=r-1\\), and the hypotheses become

\\[
1<\min(r_+,r_-),
\qquad
4<r.
\\]

With the notation of §1, ERS (1.5) becomes

\\[
\boxed{
\mathcal M_{L,O(L)}(m)
=
\prod_p\alpha_p(L,m).
}
\\]

## 3. Möbius inversion over \\(\mathbf Z\\)

Let

\\[
\iota:\langle n\rangle\longrightarrow L
\\]

be a nonzero rank-one representation. There are unique \\(d\in\mathbf Z_{>0}\\), \\(m\in\mathbf Z\setminus\{0\}\\), and a primitive embedding

\\[
\iota_0:\langle m\rangle\hookrightarrow L
\\]

such that, for generators \\(u\\) and \\(u_0\\),

\\[
\iota(u)=d\,\iota_0(u_0),
\qquad
n=d^2m.
\\]

For a primitive embedding \\(\iota:\langle n\rangle\hookrightarrow L\\), define

\\[
w_\Gamma(\iota)
:=
w_\Gamma(\iota(u)).
\\]

Replacing \\(u\\) by \\(-u\\) does not change the stabilizer, so this definition is independent of the choice of generator.

Define the primitive representation mass by

\\[
\mathcal M^{\mathrm{prim}}_{L,\Gamma}(n)
:=
\sum_{
[\iota]\in
\Gamma\backslash
\operatorname{Emb}^{\mathrm{prim}}(\langle n\rangle,L)
}
w_\Gamma(\iota).
\\]

Multiplication by \\(d\\) does not change the stabilizer of the represented line, hence

\\[
\mathcal M_{L,\Gamma}(n)
=
\sum_{d^2\mid n}
\mathcal M^{\mathrm{prim}}_{L,\Gamma}\!\left(\frac{n}{d^2}\right).
\\]

Möbius inversion gives

\\[
\boxed{
\mathcal M^{\mathrm{prim}}_{L,\Gamma}(n)
=
\sum_{d^2\mid n}
\mu(d)\,
\mathcal M_{L,\Gamma}\!\left(\frac{n}{d^2}\right).
}
\\]

## 4. The rank-one local representation density

Choose a \\(\mathbf Z\\)-basis \\(e_1,\ldots,e_r\\) of \\(L\\), and let

\\[
A
:=
\bigl(b(e_i,e_j)\bigr)_{1\le i,j\le r}.
\\]

For \\(m\ne0\\), put

\\[
X_m
:=
V(Q-m)
\subset
\mathbf A^r_{\mathbf Z}.
\\]

The rank-one specialization of Eskin--Rudnick--Sarnak (1.6) is

\\[
\boxed{
\alpha_p(L,m)
:=
\lim_{k\to\infty}
p^{-k(r-1)}
\#X_m(\mathbf Z/p^k\mathbf Z).
}
\\]

Let

\\[
S_m
:=
\left\{
p
\;\middle|\;
p\mid2m\det(A)
\right\}.
\\]

Assume \\(p\notin S_m\\). If \\(x\in X_m(\mathbf F_p)\\) were singular, then

\\[
2Ax=0
\quad\text{in }\mathbf F_p^r.
\\]

Since \\(p\nmid2\det(A)\\), this forces \\(x=0\\), contradicting \\(Q(x)=m\not\equiv0\pmod p\\). Hence \\(X_m/\mathbf Z_p\\) is smooth of relative dimension \\(r-1\\).

Let \\(x_k\in X_m(\mathbf Z/p^k\mathbf Z)\\). Every lift modulo \\(p^{k+1}\\) has the form

\\[
x_k+p^k y,
\qquad
y\in\mathbf F_p^r.
\\]

Modulo \\(p^{k+1}\\),

\\[
Q(x_k+p^k y)-m
\equiv
Q(x_k)-m
+
2p^k b(x_k,y).
\\]

After division by \\(p^k\\), the lifting condition is one nonzero affine linear equation in \\(y\in\mathbf F_p^r\\). It therefore has \\(p^{r-1}\\) solutions. Consequently

\\[
\#X_m(\mathbf Z/p^k\mathbf Z)
=
p^{(k-1)(r-1)}
\#X_m(\mathbf F_p),
\\]

and

\\[
\boxed{
\alpha_p(L,m)
=
p^{1-r}\#X_m(\mathbf F_p).
}
\\]

## 5. Hasse--Weil local zeta function at a good prime

For \\(p\notin S_m\\), define

\\[
Z_p(X_m,T)
:=
\exp\!\left(
\sum_{j\ge1}
\#X_m(\mathbf F_{p^j})\,\frac{T^j}{j}
\right).
\\]

Its logarithmic derivative satisfies

\\[
\left.
\frac{d}{dT}\log Z_p(X_m,T)
\right|_{T=0}
=
\#X_m(\mathbf F_p).
\\]

Therefore

\\[
\boxed{
\alpha_p(L,m)
=
p^{1-r}
\left.
\frac{d}{dT}\log Z_p(X_m,T)
\right|_{T=0}.
}
\\]

## 6. Good-prime point counts

Assume \\(p\notin S_m\\). Then \\(p\\) is odd and \\(p\nmid m\det(A)\\).

Let

\\[
\eta_p:\mathbf F_p^\times\longrightarrow\{1,-1\}
\\]

be the quadratic character of \\(\mathbf F_p^\times\\),

\\[
\eta_p(a):=a^{(p-1)/2}.
\\]

Equivalently,

\\[
\eta_p(a)
=
\left(\frac{a}{p}\right)
\\]

is the Legendre symbol.

R. Lidl and H. Niederreiter, *Finite Fields*, 2nd ed., Cambridge University Press, 1997, Theorems 6.26 and 6.27, give the required solution counts for nondegenerate quadratic forms over finite fields. Applying those theorems over \\(\mathbf F_{p^k}\\) yields the formulas below.

### Even rank

Suppose \\(r=2s\\), and set

\\[
D_L:=(-1)^s\det(A)\in\mathbf Z\setminus\{0\}.
\\]

Since \\(p\nmid D_L\\), the image of \\(D_L\\) in \\(\mathbf F_p^\times\\) is defined. The finite-field representation formula for a nondegenerate quadratic form gives

\\[
\#X_m(\mathbf F_{p^k})
=
p^{(2s-1)k}
-
\eta_p(D_L)^k p^{(s-1)k}.
\\]

Hence

\\[
\boxed{
Z_p(X_m,T)
=
\frac{
1-\eta_p(D_L)p^{s-1}T
}{
1-p^{2s-1}T
}
}
\\]

and

\\[
\boxed{
\alpha_p(L,m)
=
1-\eta_p(D_L)p^{-s}.
}
\\]

### Odd rank

Suppose \\(r=2s+1\\), and set

\\[
D_{L,m}:=(-1)^s m\det(A)\in\mathbf Z\setminus\{0\}.
\\]

Then \\(p\nmid D_{L,m}\\), and

\\[
\#X_m(\mathbf F_{p^k})
=
p^{2sk}
+
\eta_p(D_{L,m})^k p^{sk}.
\\]

Therefore

\\[
\boxed{
Z_p(X_m,T)
=
\frac{
1
}{
(1-p^{2s}T)
(1-\eta_p(D_{L,m})p^sT)
}
}
\\]

and

\\[
\boxed{
\alpha_p(L,m)
=
1+\eta_p(D_{L,m})p^{-s}.
}
\\]

## 7. Quadratic Dirichlet characters and the good Euler product

Let \\(D\in\mathbf Z\setminus\{0\}\\). Choose the unique squarefree integer \\(d_D\\) representing the square-class of \\(D\\) in \\(\mathbf Q^\times/\mathbf Q^{\times2}\\). Define

\\[
\Delta_D
:=
\begin{cases}
1, & d_D=1,\\
d_D, & d_D\equiv1\pmod4,\\
4d_D, & d_D\not\equiv1\pmod4.
\end{cases}
\\]

Thus, when \\(d_D\ne1\\), \\(\Delta_D\\) is the fundamental discriminant of the quadratic field

\\[
\mathbf Q(\sqrt{d_D}).
\\]

Define the quadratic Dirichlet character attached to the square-class of \\(D\\) by the Kronecker symbol

\\[
\boxed{
\chi_D(a)
:=
\left(\frac{\Delta_D}{a}\right).
}
\\]

If \\(d_D=1\\), this is the trivial Dirichlet character of conductor \\(1\\).

For every odd prime \\(p\nmid D\\),

\\[
\boxed{
\chi_D(p)
=
\eta_p(D),
}
\\]

because \\(D\\) and \\(d_D\\) determine the same square-class in \\(\mathbf F_p^\times\\).

For a finite set of primes \\(S\\), define

\\[
\zeta^S(z)
:=
\prod_{p\notin S}
(1-p^{-z})^{-1}
\\]

and

\\[
L^S(z,\chi_D)
:=
\prod_{p\notin S}
(1-\chi_D(p)p^{-z})^{-1}.
\\]

The partial Hasse--Weil zeta function of \\(X_m\\) is

\\[
\zeta^{S_m}(X_m,z)
:=
\prod_{p\notin S_m}
Z_p(X_m,p^{-z}).
\\]

If \\(r=2s\\), then \\(p\notin S_m\\) implies \\(p\nmid D_L\\), so \\(\eta_p(D_L)=\chi_{D_L}(p)\\). Hence

\\[
\boxed{
\zeta^{S_m}(X_m,z)
=
\frac{
\zeta^{S_m}(z-2s+1)
}{
L^{S_m}(z-s+1,\chi_{D_L})
}.
}
\\]

Consequently,

\\[
\boxed{
\prod_{p\notin S_m}
\alpha_p(L,m)
=
\bigl(
L^{S_m}(s,\chi_{D_L})
\bigr)^{-1}.
}
\\]

If \\(r=2s+1\\), then \\(p\notin S_m\\) implies \\(p\nmid D_{L,m}\\), so \\(\eta_p(D_{L,m})=\chi_{D_{L,m}}(p)\\). Therefore

\\[
\boxed{
\zeta^{S_m}(X_m,z)
=
\zeta^{S_m}(z-2s)\,
L^{S_m}(z-s,\chi_{D_{L,m}}),
}
\\]

and

\\[
\boxed{
\prod_{p\notin S_m}
\alpha_p(L,m)
=
\frac{
L^{S_m}(s,\chi_{D_{L,m}})
}{
\zeta^{S_m}(2s)
}.
}
\\]

## 8. Bad primes: Yang's scalar local-density formula

The explicit bad-prime formula is:

T. Yang, *An Explicit Formula for Local Densities of Quadratic Forms*, J. Number Theory **72** (1998), no. 2, 309--356, DOI 10.1006/jnth.1998.2258.

Yang defines \\(\alpha_p(T,S)\\) in equation (1.1) and the polynomial \\(\alpha(X,T,S)\\) immediately after equation (1.4), with

\\[
\alpha(1,T,S)=\alpha_p(T,S).
\\]

For the present scalar target, regard the Gram matrix \\(A\\) as a half-integral matrix over \\(\mathbf Z_p\\) and put

\\[
S_p:=A,
\qquad
T_m:=(m).
\\]

Thus

\\[
S_p[x]
=
{}^txAx
=
Q(x),
\\]

so Yang's density \\(\alpha_p(T_m,S_p)\\) is exactly \\(\alpha_p(L,m)\\).

For odd \\(p\\), Yang writes \\(S_p\\) in the diagonal form of equation (1.5),

\\[
S_p
\sim_{\mathbf Z_p}
\operatorname{diag}
\left(
\varepsilon_1p^{\ell_1},
\ldots,
\varepsilon_rp^{\ell_r}
\right),
\qquad
\varepsilon_i\in\mathbf Z_p^\times,
\qquad
\ell_1\le\cdots\le\ell_r.
\\]

For \\(t=\alpha p^a\\), with \\(\alpha\in\mathbf Z_p^\times\\), Yang defines \\(R_1(X,t,S)\\) in equation (3.2). Theorem 3.1 gives, for \\(0\le j\le1\\),

\\[
\boxed{
\alpha(X,p^j t,p^jS)
=
1+
p^jX^jR_1(X,t,S)
+
(1-p^{-1})jp^jX^j.
}
\\]

For \\(p=2\\), Yang's dyadic normal form is equation (4.1), and \\(R_1(X,t,S)\\) is defined by equations (4.3)--(4.6). Theorem 4.1 gives

\\[
\boxed{
\alpha(X,2^j t,2^jS)
=
1+
2^jX^jR_1(X,t,S)
+
jX^j,
\qquad
0\le j\le1.
}
\\]

Setting \\(X=1\\) yields the required scalar local density.

## 9. The local Poincaré series and Igusa local zeta function

For \\(p\in S_m\\), set

\\[
N_k(L,m;p)
:=
\#
\left\{
x\in L/p^kL
\;\middle|\;
Q(x)\equiv m\pmod{p^k}
\right\}
\\]

and

\\[
P_{L,m,p}(T)
:=
\sum_{k\ge0}
N_k(L,m;p)\,T^k.
\\]

Then

\\[
\alpha_p(L,m)
=
\lim_{k\to\infty}
p^{-k(r-1)}N_k(L,m;p).
\\]

For \\(f_m(x):=Q(x)-m\\), define Igusa's local zeta function by

\\[
Z_{f_m,p}(s)
:=
\int_{\mathbf Z_p^r}
|f_m(x)|_p^s\,dx.
\\]

Put \\(U=p^{-s}\\), and set

\\[
\widehat P_{L,m,p}(U)
:=
\sum_{k\ge0}
p^{-kr}N_k(L,m;p)\,U^k.
\\]

Since \\(p^{-kr}N_k(L,m;p)\\) is the normalized Haar measure of the congruence locus \\(f_m\equiv0\pmod{p^k}\\),

\\[
\boxed{
\widehat P_{L,m,p}(U)
=
\frac{
1-UZ_{f_m,p}(s)
}{
1-U
}.
}
\\]

For explicit formulas at odd \\(p\\), use:

R. A. Cowan, D. J. Katz, and L. M. White, *A New Generating Function for Calculating the Igusa Local Zeta Function*, Adv. Math. **304** (2017), 355--420, DOI 10.1016/j.aim.2016.09.003.

Their Theorem 2.1 gives the Igusa local zeta function of an arbitrary quadratic polynomial over a \\(p\\)-adic field of odd residue characteristic after the normal form described in §2.1; Proposition 4.7 gives the strong-isospectral reduction to that normal form. Applied to \\(f_m=Q-m\\), this yields \\(Z_{f_m,p}(s)\\), hence the entire sequence \\(N_k(L,m;p)\\) through the displayed relation.

## 10. The adelic genus and exact computational realizations

Let

\\[
\widehat{\mathbf Z}
:=
\prod_p\mathbf Z_p,
\qquad
\mathbf A_f
:=
\mathbf Q\otimes_{\mathbf Z}\widehat{\mathbf Z},
\qquad
\widehat L
:=
L\otimes_{\mathbf Z}\widehat{\mathbf Z}.
\\]

Put \\(V:=L_{\mathbf Q}\\) and \\(G:=O(V)\\). The stabilizer of the finite-adelic lattice is

\\[
K_L
:=
\left\{
g\in G(\mathbf A_f)
\;\middle|\;
g\widehat L=\widehat L
\right\}.
\\]

Define the class set of the genus of \\(L\\) by

\\[
\boxed{
\operatorname{Gen}(L)
:=
G(\mathbf Q)\backslash G(\mathbf A_f)/K_L.
}
\\]

If \\(g\in G(\mathbf A_f)\\), the corresponding integral lattice in \\(V\\) is

\\[
L_g
:=
V\cap g\widehat L,
\\]

with \\(V\\) embedded diagonally in \\(V\otimes_{\mathbf Q}\mathbf A_f\\). For every prime \\(p\\),

\\[
L_g\otimes\mathbf Z_p
=
g_p(L\otimes\mathbf Z_p),
\\]

so \\(L_g\\) is locally isometric to \\(L\\). Conversely, if an integral lattice \\(L'\subset V\\) is locally isometric to \\(L\\), choose \\(g_p\in G(\mathbf Q_p)\\) with \\(g_pL_p=L'_p\\). \\[

\\[

\\[

Since

\\[

=

\\]

\\[

\\[

\\[

\left(

\right).

\\[

\qquad

\\[

\qquad

Since \\(L\\) is integral,

\\[

\subset
L
\subset
M_x^\sharp.

\\[

:=
\operatorname{im}
\left(

\right).

\\[

\\[

\frac1{|O(L)_{x_1}|}
+
\frac1{|O(L)_{x_2}|}
=
\frac14+\frac1{12}
=
\frac13.
\\]

The independently computed total mass is exhausted, so

\\[
\boxed{
C^+_{L,O(L)}(2)=2.
}
\\]
