# Monomial basis of $H^0(\mathbf P^1\times\mathbf P^2,\mathcal O(4,4))$

Let $[x_0:x_1]$ be homogeneous coordinates on $\mathbf P^1$ and $[y_0:y_1:y_2]$ homogeneous coordinates on $\mathbf P^2$.

The monomial basis consists of

$$
x_0^{a_0}x_1^{a_1}y_0^{b_0}y_1^{b_1}y_2^{b_2},
\qquad a_0+a_1=4,\quad b_0+b_1+b_2=4.
$$

Hence

$$
h^0(\mathcal O(4,4))=\binom{4+1}{1}\binom{4+2}{2}=5\cdot 15=75.
$$

```python
from sage.all import QQ, PolynomialRing, show

S = PolynomialRing(QQ, names=('x0', 'x1', 'y0', 'y1', 'y2'))
x0, x1, y0, y1, y2 = S.gens()

x_monomials = [x0**a0 * x1**(4-a0) for a0 in range(5)]
y_monomials = [
    y0**b0 * y1**b1 * y2**(4-b0-b1)
    for b0 in range(5)
    for b1 in range(5-b0)
]

monomials = [mx * my for mx in x_monomials for my in y_monomials]

assert len(x_monomials) == 5
assert len(y_monomials) == 15
assert len(monomials) == 75
assert len(set(monomials)) == 75

show(monomials)
```
