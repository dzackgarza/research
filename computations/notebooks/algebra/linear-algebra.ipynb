## Matrix eigenvector scratch

```python
def matinfo(M):
    display("Matrix:")
    display(M)
    display("-------------")
    evs = [(a,b) for a,b,c in M.eigenvectors_right()]
    for l in evs:
        a, b = l
        show("lambda=", a)
        [display(bs) for bs in b]
        U = M - a * M.parent().one()
        show(U, "----->", U.rref())
        for bs in b:
            display(b)
            display(U.augment(bs).rref())
    display("-------------")
    display("Characteristic polynomial:")
    display(M.characteristic_polynomial().factor())
    display("-------------")
    J, P = M.jordan_form(transformation=True) 
    display("JCF, A = PJP^{-1}:")
    display( (P, J, P.inverse()) )
    return
```

```python
def GS(a):
    u = [None] * len(a)
    e = [None] * len(a)
    for k in range(len(a)):
        projs = [P(u[j], a[k]) for j in range(k)]
        show(k, projs)
        u[k] = (a[k] - sum(projs))
        e[k] = (u[k]).normalized()
    return (u, e)
```

```python
def printSpaces(n, nums):
    A = QQ.matrix_space(n, len(nums) // n).from_rows([nums[k * (len(nums) // n):(k + 1) * (len(nums) // n)] for k in range(n)])
    display(A)
    display(A.rref())
    display(A.transpose().rref())
    display(A.column_space())
    display(A.transpose().column_space())
    display(A.right_kernel())
    display(A.transpose().right_kernel())
    return(A)
```
